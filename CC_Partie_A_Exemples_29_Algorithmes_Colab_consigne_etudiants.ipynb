```python
# Données pédagogiques reproductibles (les étiquettes sont connues pour l'évaluation)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression, make_classification, make_blobs, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score, accuracy_score, f1_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42
X_reg, y_reg = make_regression(n_samples=280, n_features=5, noise=13, random_state=GRAINE)
X_cls, y_cls = make_classification(n_samples=400, n_features=6, n_informative=4, n_redundant=0, random_state=GRAINE)
X_clu, etiquettes_reference = make_blobs(n_samples=240, centers=3, cluster_std=0.85, random_state=GRAINE)
Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Régression:',X_reg.shape,'Classification:',X_cls.shape)
```

## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Régression linéaire (*Linear Regression*). **Famille :** régression.
**Objectif :** prédire une variable continue y à partir de variables explicatives x.
**Intuition :** on cherche l'hyperplan qui passe « au plus près » de tous les points.
**Étapes :** données (X, y) → estimation des coefficients en minimisant l'erreur quadratique → prédiction ŷ pour de nouvelles observations.
**Formule :** ŷ = w₁x₁ + … + wₚxₚ + b, avec x les variables explicatives, w les coefficients, b l'ordonnée à l'origine et ŷ la prédiction.
On minimise Σ(yᵢ − ŷᵢ)², où yᵢ est la vraie valeur de l'observation i.
**Paramètres :** w et b, appris par l'algorithme. **Hyperparamètres :** aucun dans la version simple.
**Avantages :** (1) simple et interprétable ; (2) rapide à calculer.
**Limites :** (1) suppose une relation linéaire ; (2) sensible aux valeurs extrêmes.

```python
m = LinearRegression().fit(Xr_ent[:, :1], yr_ent)
xs = np.linspace(Xr_ent[:, 0].min(), Xr_ent[:, 0].max(), 100).reshape(-1, 1)
plt.scatter(Xr_ent[:, 0], yr_ent, s=10, label="Observations")
plt.plot(xs, m.predict(xs), color="red", label="Droite ajustée")
plt.xlabel("Variable explicative x₁"); plt.ylabel("Cible y")
plt.title("Régression linéaire : la droite minimise les écarts verticaux")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque point bleu est une observation ; la droite rouge est la prédiction du modèle. L'écart vertical entre un point et la droite est l'erreur, et l'algorithme choisit la droite qui rend la somme des carrés de ces écarts minimale.

### Contextes d'utilisation
- **Cas 1 (marketing) :** prévoir les ventes mensuelles d'un produit à partir du budget publicitaire.
- **Cas 2 (immobilier / gestion) :** estimer le prix d'un logement à partir de sa surface et du nombre de pièces.
- **Données et prérequis :** variables numériques, relation approximativement linéaire, peu de colinéarité entre variables, absence de valeurs extrêmes marquées.
- **Cas peu approprié :** prédire si un client fera défaut sur un crédit (cible binaire) : la régression linéaire peut sortir des valeurs hors de [0, 1] ; la régression logistique convient mieux.
- **Métriques :** MAE, RMSE, R².

### Analyse de l'expérience Python
- **Entrées :** 5 variables explicatives synthétiques (`make_regression`), 280 observations, split 75 % / 25 %.
- **Hyperparamètres :** aucun.
- **Sortie :** MAE = 9.353512045154824.
- **Ce qu'on peut conclure :** l'erreur absolue moyenne des prédictions, exprimée dans l'unité de la cible.
- **Ce qu'on ne peut pas conclure :** les données sont synthétiques, donc cette valeur ne dit rien de la performance sur des données économiques réelles ; on ne compare pas non plus avec d'autres modèles sur une seule métrique.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Régression Ridge (*Ridge Regression*, régularisation L2). **Famille :** régression supervisée.
**Objectif :** prédire une variable continue en limitant la taille des coefficients, pour réduire le surapprentissage et gérer la colinéarité.
**Intuition :** comme la régression linéaire, mais on pénalise les coefficients trop grands, ce qui rend le modèle plus stable.
**Étapes :** standardisation des variables → minimisation de l'erreur quadratique + pénalité → prédiction.
**Formule :** on minimise Σᵢ (yᵢ − ŷᵢ)² + α Σⱼ wⱼ², avec yᵢ la vraie valeur, ŷᵢ la prédiction, wⱼ les coefficients et α ≥ 0 la force de la pénalité.
**Paramètres :** les wⱼ et b (appris). **Hyperparamètre :** α (`alpha`, fixé à 1,0 ici).
**Avantages :** (1) stable en présence de variables corrélées ; (2) limite le surapprentissage.
**Limites :** (1) ne met jamais un coefficient exactement à 0 (pas de sélection de variables) ; (2) nécessite de standardiser les variables et de choisir α.

```python
lin = make_pipeline(StandardScaler(), LinearRegression()).fit(Xr_ent, yr_ent)
alphas = [1, 100, 1000]
noms = [f"x{j+1}" for j in range(Xr_ent.shape[1])]
largeur = 0.2
pos = np.arange(len(noms))
plt.bar(pos - largeur, lin[-1].coef_, width=largeur, label="Linéaire (sans pénalité)")
for i, a in enumerate(alphas):
    r = make_pipeline(StandardScaler(), Ridge(alpha=a)).fit(Xr_ent, yr_ent)
    plt.bar(pos + i*largeur, r[-1].coef_, width=largeur, label=f"Ridge α={a}")
plt.xticks(pos, noms); plt.ylabel("Coefficient (variables standardisées)")
plt.title("Ridge : plus α est grand, plus les coefficients sont réduits")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque groupe de barres correspond à une variable (x1 à x5). La première barre est le coefficient de la régression linéaire sans pénalité ; les suivantes sont ceux de Ridge pour α = 1, 100 et 1000. Plus α augmente, plus les barres rétrécissent vers 0, sans jamais s'annuler exactement. Les variables standardisées permettent de comparer les coefficients entre eux.

### Contextes d'utilisation
- **Cas 1 (finance) :** prévoir le rendement d'un portefeuille à partir de nombreux indicateurs de marché fortement corrélés entre eux.
- **Cas 2 (marketing) :** estimer la demande d'un produit à partir de nombreuses variables (prix, promotions, saison, canaux de vente) souvent corrélées.
- **Données et prérequis :** variables numériques, standardisées avant l'entraînement ; α à choisir par validation croisée.
- **Cas peu approprié :** quand on veut identifier un petit nombre de variables réellement utiles : Ridge garde toutes les variables, donc Lasso conviendrait mieux.
- **Métriques :** MAE, RMSE, R².

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que l'algorithme 1 (210 observations d'entraînement, 70 de test, 5 variables), standardisées via `StandardScaler` dans un pipeline.
- **Hyperparamètres :** `alpha=1.0`.
- **Sortie :** MAE ≈ 9,37 sur le jeu de test, contre ≈ 9,35 pour la régression linéaire.
- **Ce qu'on peut conclure :** avec α = 1, la pénalité est très faible et les deux modèles se comportent presque de la même façon sur ces données. L'écart de MAE (environ 0,01) est très petit.
- **Ce qu'on ne peut pas conclure :** que Ridge est moins bon que la régression linéaire. Une différence aussi faible sur un seul découpage ne permet pas de les départager. Les données sont synthétiques, et α n'a pas été optimisé (par exemple par validation croisée), donc on ne sait pas quel α donnerait le meilleur résultat.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Régression Lasso (*Least Absolute Shrinkage and Selection Operator*, régularisation L1). **Famille :** régression supervisée.
**Objectif :** prédire une variable continue tout en sélectionnant automatiquement les variables utiles, en mettant à 0 les coefficients des variables peu informatives.
**Intuition :** comme Ridge, on pénalise les coefficients, mais avec une pénalité en valeur absolue qui peut les annuler exactement.
**Étapes :** standardisation des variables → minimisation de l'erreur quadratique + pénalité L1 → prédiction.
**Formule :** on minimise (1/(2n)) Σᵢ (yᵢ − ŷᵢ)² + α Σⱼ |wⱼ|, avec n le nombre d'observations, yᵢ la vraie valeur, ŷᵢ la prédiction, wⱼ les coefficients et α ≥ 0 la force de la pénalité.
**Paramètres :** les wⱼ et b (appris). **Hyperparamètres :** α (`alpha`, fixé à 0,1 ici) et `max_iter` (nombre maximal d'itérations de l'algorithme de résolution).
**Avantages :** (1) sélectionne des variables et produit un modèle parcimonieux ; (2) plus facile à interpréter.
**Limites :** (1) instable quand des variables sont très corrélées (il en retient une au hasard) ; (2) nécessite de standardiser les variables et de choisir α.

```python
alphas = [0.1, 1, 5, 20, 50]
noms = [f"x{j+1}" for j in range(Xr_ent.shape[1])]
coefs = []
for a in alphas:
    l = make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=10000)).fit(Xr_ent, yr_ent)
    coefs.append(l[-1].coef_)
coefs = np.array(coefs)
for j in range(coefs.shape[1]):
    plt.plot(alphas, coefs[:, j], marker="o", label=noms[j])
plt.xscale("log"); plt.xlabel("α (échelle logarithmique)")
plt.ylabel("Coefficient (variables standardisées)")
plt.title("Lasso : quand α augmente, des coefficients tombent à 0")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque courbe représente le coefficient d'une variable (x1 à x5) en fonction de α. À gauche (α petit), le modèle ressemble à la régression linéaire ; en allant vers la droite, les coefficients diminuent. Les courbes qui atteignent 0 correspondent aux variables que Lasso élimine : x1 et x2 s'annulent dès α = 20, x3 à α = 50. Les variables x4 et x5, les plus influentes, restent non nulles plus longtemps.

### Contextes d'utilisation
- **Cas 1 (finance) :** parmi des dizaines d'indicateurs financiers, retenir ceux qui expliquent réellement le score de risque d'une entreprise.
- **Cas 2 (marketing) :** identifier quelles variables (canaux, promotions, caractéristiques clients) influencent vraiment les ventes.
- **Données et prérequis :** variables numériques standardisées ; α à choisir par validation croisée.
- **Cas peu approprié :** quand toutes les variables sont corrélées et utiles ensemble : Lasso en retiendrait arbitrairement quelques-unes, et Ridge serait plus stable.
- **Métriques :** MAE, RMSE, R², ainsi que le nombre de coefficients non nuls.

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que précédemment (210 observations d'entraînement, 70 de test, 5 variables), standardisées via `StandardScaler`.
- **Hyperparamètres :** `alpha=0.1`, `max_iter=10000`.
- **Sortie :** R² ≈ 0,986 sur le jeu de test, c'est-à-dire que le modèle explique environ 98,6 % de la variance de la cible.
- **Ce qu'on peut conclure :** avec α = 0,1, la pénalité est faible, aucun coefficient ne s'annule et le modèle se comporte presque comme la régression linéaire (R² ≈ 0,986 aussi).
- **Ce qu'on ne peut pas conclure :** que Lasso est meilleur ou moins bon que la régression linéaire sur ces données. Les données sont synthétiques, α n'est pas optimisé, et on ne voit pas ici l'effet de sélection de variables, qui apparaît seulement pour des α plus grands (voir le schéma).

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Régression polynomiale (*Polynomial Regression*). **Famille :** régression supervisée.
**Objectif :** modéliser une relation non linéaire entre x et y en restant dans le cadre d'un modèle linéaire.
**Intuition :** on ajoute aux variables d'origine leurs puissances (x², x³…) et leurs produits croisés (x₁x₂…), puis on ajuste une régression linéaire sur ces nouvelles variables.
**Étapes :** standardisation → création des variables polynomiales de degré d → régression (ici Ridge) → prédiction.
**Formule (une variable) :** ŷ = w₀ + w₁x + w₂x² + … + w_d x^d, avec x la variable explicative, wₖ les coefficients et d le degré du polynôme.
**Paramètres :** les coefficients wₖ (appris). **Hyperparamètres :** le degré d (`degree=2`) et la pénalité α de Ridge (`alpha=10`).
**Avantages :** (1) capture des courbes et des interactions entre variables ; (2) reste simple à ajuster.
**Limites :** (1) surapprentissage rapide quand le degré augmente ; (2) le nombre de variables explose avec le degré et le nombre de variables d'origine.

```python
rng = np.random.default_rng(GRAINE)
x = np.sort(rng.uniform(-3, 3, 60))
y = 0.5 * x**2 - x + rng.normal(0, 0.8, 60)
xs = np.linspace(-3, 3, 300).reshape(-1, 1)
plt.scatter(x, y, s=12, label="Observations")
for d, couleur in [(1, "green"), (2, "red"), (10, "purple")]:
    p = make_pipeline(PolynomialFeatures(d), LinearRegression()).fit(x.reshape(-1, 1), y)
    plt.plot(xs, p.predict(xs), color=couleur, label=f"Degré {d}")
plt.ylim(y.min() - 1, y.max() + 1)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Régression polynomiale : sous-apprentissage (d=1) et surapprentissage (d=10)")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** les points suivent une courbe en U. La droite verte (degré 1) est trop rigide : elle ne suit pas la courbure (sous-apprentissage). La courbe rouge (degré 2) épouse bien la forme générale. La courbe violette (degré 10) est très flexible et peut osciller pour suivre le bruit, surtout aux extrémités (risque de surapprentissage). *Ces données sont générées uniquement pour illustrer le principe ; elles ne proviennent pas du jeu de données utilisé dans le code de l'exemple.*

### Contextes d'utilisation
- **Cas 1 (marketing) :** modéliser l'effet d'une dépense publicitaire sur les ventes, avec des rendements décroissants (relation courbe plutôt que droite).
- **Cas 2 (logistique) :** estimer le coût de transport en fonction de la distance, quand le coût ne croît pas linéairement.
- **Données et prérequis :** peu de variables explicatives, variables standardisées, choix du degré par validation croisée.
- **Cas peu approprié :** un grand nombre de variables d'origine (par exemple 100) : le degré 2 crée alors des milliers de variables, d'où surapprentissage et temps de calcul élevé.
- **Métriques :** MAE, RMSE, R², comparées sur des données de test.

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que précédemment (210 observations d'entraînement, 70 de test, 5 variables). Le degré 2 transforme les 5 variables en 20 variables (5 variables + 5 carrés + 10 produits croisés).
- **Hyperparamètres :** `degree=2`, `alpha=10` (Ridge), `include_bias=False`.
- **Sortie :** MAE ≈ 10,90 sur le jeu de test, contre ≈ 9,35 pour la régression linéaire.
- **Ce qu'on peut conclure :** sur ces données, le modèle polynomial de degré 2 est un peu moins précis que le modèle linéaire. Cela est cohérent avec la façon dont les données sont générées (`make_regression` produit une relation linéaire) : les termes quadratiques n'apportent pas d'information utile et ajoutent surtout du bruit.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Arbre de régression (*Regression Tree*, CART). **Famille :** régression supervisée.
**Objectif :** prédire une variable continue en découpant l'espace des variables en régions, chacune associée à une valeur prédite.
**Intuition :** l'arbre pose une suite de questions du type « x_j ≤ seuil ? » ; chaque réponse mène à une branche, jusqu'à une feuille qui donne la prédiction.
**Étapes :** à chaque nœud, choix de la variable et du seuil qui réduisent le plus l'erreur → division des données en deux groupes → répétition jusqu'à la profondeur maximale → prédiction = moyenne de y dans la feuille atteinte.
**Critère :** à chaque division, on minimise Σ(yᵢ − ȳ_gauche)² + Σ(yᵢ − ȳ_droite)², avec yᵢ les valeurs de la cible et ȳ la moyenne de chaque groupe.
**Paramètres :** variables et seuils de découpage, valeurs des feuilles (appris). **Hyperparamètres :** profondeur maximale (`max_depth=5`), nombre minimal d'observations par feuille, etc.
**Avantages :** (1) facile à interpréter et à visualiser ; (2) capture des relations non linéaires sans standardisation.
**Limites :** (1) surapprentissage si l'arbre est profond ; (2) prédictions en paliers, peu adaptées aux relations linéaires lisses.

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree
arbre_visu = DecisionTreeRegressor(max_depth=2, random_state=GRAINE).fit(Xr_ent, yr_ent)
plt.figure(figsize=(10, 5))
plot_tree(arbre_visu, feature_names=[f"x{j+1}" for j in range(Xr_ent.shape[1])], filled=True, rounded=True)
plt.title("Arbre de régression (profondeur 2) : chaque feuille donne une valeur prédite")
plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque case est un nœud. Pour une nouvelle observation, on part du haut et on répond à la question écrite dans la case : si c'est vrai, on va à gauche, sinon à droite. On continue jusqu'à une feuille (case sans question) : `value` est la prédiction (moyenne de y dans cette feuille), `samples` le nombre d'observations qui y arrivent et `squared_error` l'erreur quadratique moyenne à ce nœud. Les couleurs reflètent la valeur prédite. Les questions portent ici surtout sur x5 et x4, les variables les plus influentes. *Le schéma montre un arbre de profondeur 2 pour la lisibilité ; le modèle de l'exemple a une profondeur 5.*

### Contextes d'utilisation
- **Cas 1 (logistique) :** estimer le délai de livraison selon la distance, le poids et le mode de transport, avec des règles de décision simples à expliquer.
- **Cas 2 (finance / gestion) :** estimer le montant d'un prêt à accorder selon le revenu, l'âge et l'ancienneté, avec des seuils lisibles par un conseiller.
- **Données et prérequis :** variables numériques ou catégorielles encodées ; pas de standardisation nécessaire ; profondeur à limiter.
- **Cas peu approprié :** une relation linéaire lisse entre variables (comme dans les données de cet exemple) : l'arbre la découpe en paliers et prédit moins bien qu'une régression linéaire.
- **Métriques :** MAE, RMSE, R², comparées entre entraînement et test pour repérer le surapprentissage.

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que précédemment (210 observations d'entraînement, 70 de test, 5 variables). Aucune standardisation n'est nécessaire pour un arbre.
- **Hyperparamètres :** `max_depth=5`, `random_state=42`.
- **Sortie :** MAE ≈ 37,72 sur le jeu de test, contre ≈ 9,35 pour la régression linéaire.
- **Ce qu'on peut conclure :** sur ces données, l'arbre de profondeur 5 est nettement moins précis que les modèles linéaires. Les données ayant été générées avec une relation linéaire, l'arbre doit l'approcher par des paliers constants, ce qui donne une erreur élevée.
- **Ce qu'on ne peut pas conclure :** que les arbres de régression sont peu performants en général. Ils peuvent être très bons sur des relations non linéaires ou avec des interactions. De plus, la profondeur n'a pas été optimisée et le résultat provient d'un seul découpage de données synthétiques.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Forêt aléatoire pour la régression (*Random Forest Regressor*). **Famille :** régression supervisée, méthode d'ensemble (bagging).
**Objectif :** prédire une variable continue en moyennant les prédictions de nombreux arbres, pour réduire l'instabilité d'un arbre seul.
**Intuition :** chaque arbre est entraîné sur un échantillon différent des données (tirage avec remise) et ne considère à chaque division qu'un sous-ensemble aléatoire de variables ; on moyenne leurs prédictions.
**Étapes :** tirage de B échantillons bootstrap → entraînement d'un arbre par échantillon → prédiction = moyenne des B arbres.
**Formule :** ŷ(x) = (1/B) Σ_{b=1..B} T_b(x), avec B le nombre d'arbres, T_b le b-ième arbre et x l'observation à prédire.
**Paramètres :** les découpages et valeurs de feuilles de chaque arbre (appris). **Hyperparamètres :** nombre d'arbres (`n_estimators=80`), profondeur maximale, nombre de variables testées à chaque division, graine (`random_state=42`).
**Avantages :** (1) bien plus stable et moins sujette au surapprentissage qu'un arbre seul ; (2) gère les non-linéarités et fournit l'importance des variables.
**Limites :** (1) moins interprétable qu'un arbre unique ; (2) prédictions en paliers, donc peu adaptée aux relations linéaires lisses ; elle ne prédit pas hors de la plage des valeurs d'entraînement.

```python
foret = RandomForestRegressor(n_estimators=80, random_state=GRAINE).fit(Xr_ent, yr_ent)
noms = [f"x{j+1}" for j in range(Xr_ent.shape[1])]
plt.bar(noms, foret.feature_importances_, color="steelblue")
plt.ylabel("Importance (somme = 1)")
plt.title("Forêt aléatoire : importance des variables (régression)")
plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque barre indique la part de la réduction d'erreur due à une variable, moyennée sur les 80 arbres ; la somme vaut 1. Plus la barre est haute, plus la variable est utilisée dans les découpages importants. Ici x4 et x5 dominent (environ 0,44 et 0,47), ce qui correspond aux deux coefficients les plus grands de la régression linéaire. *L'importance mesure l'utilité d'une variable pour le modèle, pas un effet causal.*

### Contextes d'utilisation
- **Cas 1 (commerce / logistique) :** prévoir la demande d'un produit en magasin à partir du prix, de la saison, des promotions et de la météo, avec des relations non linéaires.
- **Cas 2 (finance) :** estimer le prix d'un bien immobilier ou d'un actif à partir de caractéristiques mélangeant variables numériques et catégorielles.
- **Données et prérequis :** variables numériques ou catégorielles encodées ; pas de standardisation nécessaire ; assez d'observations pour que les arbres soient variés.
- **Cas peu approprié :** une relation linéaire lisse ou une extrapolation hors de la plage des données (comme dans cet exemple) : la forêt prédit par paliers et ne dépasse jamais les valeurs vues à l'entraînement.
- **Métriques :** MAE, RMSE, R², importance des variables, erreur out-of-bag.

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que précédemment (210 observations d'entraînement, 70 de test, 5 variables).
- **Hyperparamètres :** `n_estimators=80`, `random_state=42`, autres valeurs par défaut.
- **Sortie :** MAE ≈ 24,37 sur le jeu de test, contre ≈ 37,72 pour l'arbre de régression (profondeur 5) et ≈ 9,35 pour la régression linéaire.
- **Ce qu'on peut conclure :** moyenner plusieurs arbres réduit nettement l'erreur par rapport à un arbre seul. En revanche, sur ces données générées avec une relation linéaire, les modèles linéaires restent meilleurs.
- **Ce qu'on ne peut pas conclure :** que la forêt est moins bonne en général. Sur des données réelles non linéaires, elle est souvent compétitive. Les hyperparamètres n'ont pas été optimisés et le résultat provient d'un seul découpage de données synthétiques.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation
**Nom :** Gradient Boosting pour la régression (*Gradient Boosting Regressor*, ici la version par histogrammes `HistGradientBoostingRegressor`). **Famille :** régression supervisée, méthode d'ensemble (boosting).
**Objectif :** prédire une variable continue en construisant les arbres l'un après l'autre, chacun corrigeant les erreurs du précédent.
**Intuition :** on part d'une prédiction simple (la moyenne), on calcule les erreurs restantes (résidus), on entraîne un petit arbre pour les prédire et on l'ajoute au modèle avec un pas réduit ; on répète.
**Étapes :** prédiction initiale → calcul des résidus → ajustement d'un arbre sur les résidus → mise à jour du modèle → répétition M fois.
**Formule :** F_m(x) = F_{m−1}(x) + η · h_m(x), avec F_m le modèle après m étapes, h_m le m-ième arbre entraîné sur les résidus et η le taux d'apprentissage (`learning_rate`).
**Paramètres :** les arbres h_m (appris). **Hyperparamètres :** nombre d'itérations (`max_iter=80`), taux d'apprentissage η, profondeur des arbres, graine (`random_state=42`).
**Avantages :** (1) très performant sur des données tabulaires complexes ; (2) gère les non-linéarités et les interactions.
**Limites :** (1) sensible au surapprentissage si trop d'itérations ; (2) plusieurs hyperparamètres à régler et moins interprétable.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
iterations = [1, 5, 10, 20, 40, 80, 150, 300]
err_ent, err_test = [], []
for n in iterations:
    r = HistGradientBoostingRegressor(max_iter=n, random_state=GRAINE).fit(Xr_ent, yr_ent)
    err_ent.append(mean_absolute_error(yr_ent, r.predict(Xr_ent)))
    err_test.append(mean_absolute_error(yr_test, r.predict(Xr_test)))
plt.plot(iterations, err_ent, marker="o", label="Erreur entraînement")
plt.plot(iterations, err_test, marker="o", label="Erreur test")
plt.xscale("log"); plt.xlabel("Nombre d'itérations (arbres)"); plt.ylabel("MAE")
plt.title("Gradient Boosting : l'erreur baisse à chaque arbre ajouté")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** l'axe horizontal donne le nombre d'arbres (échelle logarithmique), l'axe vertical la MAE. Les deux courbes baissent quand on ajoute des arbres : chaque arbre corrige une partie de l'erreur restante. L'erreur d'entraînement continue de diminuer, tandis que l'erreur de test se stabilise autour de 22 à 23 après environ 80 à 150 arbres, puis remonte légèrement : l'écart qui se creuse entre les deux courbes est le signe d'un début de surapprentissage.

### Contextes d'utilisation
- **Cas 1 (finance) :** prévoir le montant de pertes sur un portefeuille de crédits à partir de nombreuses caractéristiques clients.
- **Cas 2 (commerce / logistique) :** prévoir les ventes ou les délais de livraison à partir de variables mixtes (prix, saison, distance, canal) avec des interactions complexes.
- **Données et prérequis :** données tabulaires, de taille moyenne à grande ; pas de standardisation nécessaire ; hyperparamètres à régler avec une validation séparée.
- **Cas peu approprié :** très peu de données ou relation linéaire simple (comme dans cet exemple) : un modèle linéaire est plus précis, plus simple et plus stable.
- **Métriques :** MAE, RMSE, R², courbes d'erreur entraînement / validation.

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que précédemment (210 observations d'entraînement, 70 de test, 5 variables).
- **Hyperparamètres :** `max_iter=80`, `random_state=42`, autres valeurs par défaut (dont le taux d'apprentissage de 0,1).
- **Sortie :** MAE ≈ 23,27 sur le jeu de test, contre ≈ 24,37 pour la forêt aléatoire, ≈ 37,72 pour l'arbre seul et ≈ 9,35 pour la régression linéaire.
- **Ce qu'on peut conclure :** le boosting fait mieux qu'un arbre seul et est du même ordre que la forêt aléatoire. Sur ces données générées avec une relation linéaire, les modèles linéaires restent bien meilleurs.
- **Ce qu'on ne peut pas conclure :** que le gradient boosting est moins bon que la forêt ou la régression linéaire en général. L'écart avec la forêt est faible, le nombre d'itérations et les autres hyperparamètres n'ont pas été optimisés, et les données sont synthétiques avec un seul découpage.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
**Nom :** Régression logistique (*Logistic Regression*). **Famille :** classification supervisée (binaire ici).
**Objectif :** estimer la probabilité qu'une observation appartienne à la classe 1, puis la classer selon un seuil (0,5 par défaut).
**Intuition :** on calcule un score linéaire à partir des variables, puis on le transforme en probabilité grâce à la fonction sigmoïde, qui écrase toute valeur dans l'intervalle [0, 1].
**Étapes :** standardisation → estimation des coefficients par maximum de vraisemblance → calcul de P(y=1|x) → classe 1 si P ≥ 0,5, sinon classe 0.
**Formule :** P(y=1|x) = 1 / (1 + e^{−(w·x + b)}), avec x les variables explicatives, w les coefficients, b l'ordonnée à l'origine et e^{−z} l'exponentielle. On minimise la perte logistique −Σ[yᵢ log pᵢ + (1−yᵢ) log(1−pᵢ)], où pᵢ est la probabilité prédite pour l'observation i.
**Paramètres :** w et b (appris). **Hyperparamètres :** force de régularisation (`C`, 1,0 par défaut), nombre maximal d'itérations (`max_iter=1000`), seuil de décision.
**Avantages :** (1) donne des probabilités et des coefficients interprétables ; (2) rapide et stable.
**Limites :** (1) frontière de décision linéaire seulement ; (2) sensible à la colinéarité et nécessite de standardiser les variables.

```python
cols = [2, 5]  # deux variables, pour pouvoir tracer en 2D
m2 = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, random_state=GRAINE)).fit(Xc_ent[:, cols], yc_ent)
g1, g2 = np.meshgrid(np.linspace(Xc_ent[:, 2].min()-1, Xc_ent[:, 2].max()+1, 300),
                     np.linspace(Xc_ent[:, 5].min()-1, Xc_ent[:, 5].max()+1, 300))
P = m2.predict_proba(np.c_[g1.ravel(), g2.ravel()])[:, 1].reshape(g1.shape)
plt.contourf(g1, g2, P, levels=20, cmap="coolwarm", alpha=0.6)
plt.colorbar(label="P(classe 1)")
plt.contour(g1, g2, P, levels=[0.5], colors="k")
plt.scatter(Xc_ent[:, 2], Xc_ent[:, 5], c=yc_ent, cmap="coolwarm", edgecolor="k", s=15)
plt.xlabel("Variable x3"); plt.ylabel("Variable x6")
plt.title("Régression logistique : frontière linéaire (trait noir, P = 0,5)")
plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque point est une observation d'entraînement, coloré selon sa classe (0 ou 1). Le fond coloré indique la probabilité prédite d'appartenir à la classe 1, de bleu (proche de 0) à rouge (proche de 1). Le trait noir est la frontière de décision (P = 0,5) : c'est une droite, car le modèle est linéaire. Plus un point est loin de la frontière, plus la probabilité est proche de 0 ou de 1. *Le schéma n'utilise que 2 des 6 variables ; le modèle du code en utilise 6, d'où quelques points mal classés sur le graphique.*

### Contextes d'utilisation
- **Cas 1 (finance) :** estimer la probabilité de défaut de paiement d'un client à partir de son revenu, de son endettement et de son historique de crédit.
- **Cas 2 (marketing) :** prédire si un client va résilier son abonnement (*churn*) ou répondre à une campagne.
- **Données et prérequis :** variables numériques standardisées (ou catégorielles encodées), cible binaire (ou multiclasse), peu de colinéarité.
- **Cas peu approprié :** une frontière très non linéaire entre les classes (par exemple une classe au centre entourée par l'autre) : un modèle à frontière courbe, comme un SVM à noyau ou un arbre, convient mieux.
- **Métriques :** F1-score, précision, rappel, ROC-AUC ou PR-AUC (surtout en cas de classes déséquilibrées), matrice de confusion.

### Analyse de l'expérience Python
- **Entrées :** 400 observations synthétiques (`make_classification`, 6 variables dont 4 informatives, 2 classes), séparées en 300 pour l'entraînement et 100 pour le test de façon stratifiée (`stratify`), donc 50 observations de chaque classe en test.
- **Hyperparamètres :** `max_iter=1000`, `random_state=42`, régularisation par défaut (`C=1`).
- **Sortie :** F1 ≈ 0,81 sur le jeu de test.
- **Ce qu'on peut conclure :** le modèle classe correctement une bonne partie des observations, avec un compromis entre précision et rappel d'environ 0,81 pour la classe 1. Les classes étant équilibrées, ce score est facile à interpréter.
- **Ce qu'on ne peut pas conclure :** que ce modèle est bon ou mauvais en général : on n'a ni comparaison avec d'autres algorithmes, ni analyse séparée de la précision et du rappel, ni ROC-AUC. Les données sont synthétiques et un seul découpage est utilisé ; sur 100 observations de test, le score peut varier sensiblement d'un découpage à l'autre.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
**Nom :** K plus proches voisins (*k-Nearest Neighbors*, KNN). **Famille :** classification supervisée (utilisable aussi en régression).
**Objectif :** classer une observation d'après la classe majoritaire de ses k voisins les plus proches dans l'ensemble d'entraînement.
**Intuition :** « dis-moi qui sont tes voisins, je te dirai qui tu es » : deux observations proches ont de bonnes chances d'appartenir à la même classe.
**Étapes :** standardisation des variables → calcul des distances entre la nouvelle observation et chaque observation d'entraînement → sélection des k plus proches → vote majoritaire → classe prédite. Il n'y a pas de véritable phase d'apprentissage : les données sont simplement mémorisées.
**Formule :** distance euclidienne d(x, x') = √(Σⱼ (xⱼ − x'ⱼ)²), avec x l'observation à classer, x' une observation d'entraînement et j l'indice de la variable ; ŷ = classe la plus fréquente parmi les k voisins.
**Paramètres :** aucun paramètre appris (modèle basé sur les données). **Hyperparamètres :** k (`n_neighbors=5`), la distance, la pondération des voisins.
**Avantages :** (1) très simple et sans hypothèse sur la forme de la frontière ; (2) capte des frontières non linéaires.
**Limites :** (1) sensible à l'échelle des variables et à la dimension ; (2) prédiction lente sur de grands jeux de données, car il faut calculer toutes les distances.

```python
from sklearn.neighbors import KNeighborsClassifier
cols = [2, 5]  # deux variables pour tracer en 2D
g1, g2 = np.meshgrid(np.linspace(Xc_ent[:, 2].min()-1, Xc_ent[:, 2].max()+1, 300),
                     np.linspace(Xc_ent[:, 5].min()-1, Xc_ent[:, 5].max()+1, 300))
fig, axs = plt.subplots(1, 2, figsize=(11, 4))
for ax, k in zip(axs, [1, 15]):
    mk = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k)).fit(Xc_ent[:, cols], yc_ent)
    Z = mk.predict(np.c_[g1.ravel(), g2.ravel()]).reshape(g1.shape)
    ax.contourf(g1, g2, Z, alpha=0.4, cmap="coolwarm")
    ax.scatter(Xc_ent[:, 2], Xc_ent[:, 5], c=yc_ent, cmap="coolwarm", edgecolor="k", s=15)
    ax.set_title(f"KNN avec k = {k}")
    ax.set_xlabel("Variable x3"); ax.set_ylabel("Variable x6")
plt.suptitle("KNN : plus k est petit, plus la frontière est irrégulière")
plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque point est une observation d'entraînement, colorée selon sa classe. Le fond indique la classe que KNN prédirait pour un nouveau point situé à cet endroit. À gauche (k = 1), la frontière est très irrégulière et suit chaque point isolé : le modèle est très flexible et risque de surapprendre le bruit. À droite (k = 15), la frontière est plus lisse : le modèle est plus stable mais peut manquer des détails locaux. *Le schéma utilise seulement 2 des 6 variables pour pouvoir tracer en 2D.*

### Contextes d'utilisation
- **Cas 1 (commerce / marketing) :** recommander des produits ou classer un client dans un segment d'après les clients les plus similaires.
- **Cas 2 (finance) :** classer une demande de crédit d'après les dossiers passés les plus proches (revenu, âge, endettement).
- **Données et prérequis :** variables numériques standardisées, nombre de variables modéré, jeu d'entraînement conservé en mémoire.
- **Cas peu approprié :** un très grand jeu de données ou un grand nombre de variables (haute dimension) : les distances deviennent peu informatives et la prédiction devient lente.
- **Métriques :** F1-score, précision, rappel, ROC-AUC ; k choisi par validation croisée.

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que l'algorithme 8 (300 observations d'entraînement, 100 de test, 6 variables, 2 classes), standardisées via `StandardScaler`.
- **Hyperparamètres :** `n_neighbors=5`, autres valeurs par défaut (distance euclidienne, poids uniformes).

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
**Nom :** Naive Bayes gaussien (*Gaussian Naive Bayes*). **Famille :** classification supervisée probabiliste.
**Objectif :** classer une observation en calculant la probabilité de chaque classe à partir des valeurs de ses variables, grâce au théorème de Bayes.
**Intuition :** pour chaque classe, on apprend la distribution de chaque variable (une gaussienne ici) ; on combine ensuite ces distributions en supposant, de façon « naïve », que les variables sont indépendantes entre elles à l'intérieur d'une classe.
**Étapes :** estimation de la probabilité a priori de chaque classe et de la moyenne et de la variance de chaque variable par classe → calcul de la probabilité a posteriori de chaque classe pour une nouvelle observation → choix de la classe la plus probable.
**Formule :** P(c | x) ∝ P(c) · Πⱼ P(xⱼ | c), avec c la classe, x = (x₁, …, x_p) les variables, P(c) la probabilité a priori de la classe et P(xⱼ | c) la densité gaussienne de la variable j dans la classe c, définie par sa moyenne μ_{j,c} et son écart-type σ_{j,c}.
**Paramètres :** les P(c), μ_{j,c} et σ_{j,c} (appris). **Hyperparamètres :** très peu ; ici, le lissage de variance (`var_smoothing`, valeur par défaut).
**Avantages :** (1) très rapide, même avec peu de données ; (2) fournit des probabilités et peut gérer beaucoup de variables.
**Limites :** (1) l'hypothèse d'indépendance des variables est rarement vraie ; (2) l'hypothèse gaussienne peut être inadaptée, et les probabilités sont souvent mal calibrées.

```python
from sklearn.naive_bayes import GaussianNB
from scipy.stats import norm
nb = GaussianNB().fit(Xc_ent, yc_ent)
j = 5  # variable x6
xs = np.linspace(Xc_ent[:, j].min(), Xc_ent[:, j].max(), 200)
couleurs = ["tab:blue", "tab:red"]
for c in [0, 1]:
    plt.hist(Xc_ent[yc_ent == c, j], bins=20, density=True, alpha=0.35, color=couleurs[c], label=f"Données classe {c}")
    plt.plot(xs, norm.pdf(xs, nb.theta_[c, j], np.sqrt(nb.var_[c, j])), color=couleurs[c], label=f"Gaussienne apprise classe {c}")
plt.xlabel("Variable x6"); plt.ylabel("Densité")
plt.title("Naive Bayes : une gaussienne par classe pour la variable x6")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** les histogrammes montrent la distribution réelle de la variable x6 dans chaque classe ; les courbes sont les gaussiennes apprises par le modèle (moyenne et écart-type estimés pour chaque classe). Pour une nouvelle observation, Naive Bayes lit la hauteur de chaque courbe à la valeur observée : plus la courbe d'une classe est haute, plus cette classe est vraisemblable. Il fait la même chose pour chaque variable, puis multiplie les résultats avec la probabilité a priori de la classe. *Le schéma ne montre qu'une des 6 variables ; la décision finale combine les 6.*

### Contextes d'utilisation
- **Cas 1 (marketing / gestion) :** filtrer les courriels indésirables ou classer des avis clients selon leur sentiment à partir des mots employés (version multinomiale).

- **Données et prérequis :** variables supposées indépendantes dans chaque classe ; pour la version gaussienne, variables numériques à peu près symétriques ; peu de données suffisent.
- **Cas peu approprié :** variables fortement corrélées (par exemple plusieurs indicateurs financiers qui mesurent la même chose) : l'hypothèse d'indépendance est violée et les probabilités deviennent peu fiables.
- **Métriques :** F1-score, précision, rappel, ROC-AUC ; vérifier aussi la calibration des probabilités.

- **Hyperparamètres :** valeurs par défaut (`var_smoothing=1e-9`) ; aucune standardisation n'est nécessaire.
- **Sortie :** F1 ≈ 0,76 sur le jeu de test, contre ≈ 0,81 pour la régression logistique et ≈ 0,89 pour KNN.

- **Ce qu'on ne peut pas conclure :** que Naive Bayes est mauvais en général, ni que cette explication est prouvée : on ne l'a pas testée ici. Les données sont synthétiques, le test ne contient que 100 observations, et un seul découpage est utilisé.

### Présentation
**Nom :** Arbre de décision (*Decision Tree Classifier*, CART). **Famille :** classification supervisée.
**Objectif :** classer une observation grâce à une suite de questions simples sur ses variables.
**Intuition :** l'arbre pose des questions du type « x_j ≤ seuil ? » ; chaque réponse mène à une branche, jusqu'à une feuille qui donne la classe prédite.
**Étapes :** à chaque nœud, choix de la variable et du seuil qui séparent le mieux les classes → division des données en deux groupes → répétition jusqu'à la profondeur maximale → prédiction = classe majoritaire de la feuille atteinte.
**Critère :** à chaque division, on minimise l'impureté de Gini, G = 1 − Σ_k p_k², avec p_k la proportion d'observations de la classe k dans le nœud. G vaut 0 quand le nœud ne contient qu'une seule classe.
**Paramètres :** variables et seuils de découpage, classes des feuilles (appris). **Hyperparamètres :** profondeur maximale (`max_depth=4`), nombre minimal d'observations par feuille, critère de division.
**Avantages :** (1) très lisible et facile à expliquer à un non-spécialiste ; (2) pas besoin de standardiser et capte des relations non linéaires.
**Limites :** (1) surapprentissage si l'arbre est profond ; (2) instable : un léger changement des données peut modifier tout l'arbre.

Dans chaque case, `gini` mesure le mélange des classes (0 = nœud pur), `samples` est le nombre d'observations qui y arrivent, `value` le nombre d'observations de chaque classe, et `class` la classe prédite. La couleur reflète la classe majoritaire : plus elle est foncée, plus le nœud est pur. Ici, la première question porte sur x6, puis sur x1 ou x3 selon la branche. *Le schéma montre une profondeur de 2 pour la lisibilité ; le modèle de l'exemple a une profondeur de 4.*

### Présentation
**Nom :** Forêt aléatoire pour la classification (*Random Forest Classifier*). **Famille :** classification supervisée, méthode d'ensemble (bagging).
**Objectif :** classer une observation par un vote de nombreux arbres de décision, pour obtenir un modèle plus stable et plus précis qu'un arbre seul.
**Intuition :** chaque arbre est entraîné sur un échantillon différent des données (tirage avec remise) et ne considère à chaque division qu'un sous-ensemble aléatoire de variables ; les arbres sont ainsi variés et leurs erreurs se compensent en partie.

**Paramètres :** découpages et feuilles de chaque arbre (appris). **Hyperparamètres :** nombre d'arbres (`n_estimators=80`), profondeur maximale, nombre de variables testées à chaque division, graine (`random_state=42`).
**Avantages :** (1) bien plus stable et moins sujette au surapprentissage qu'un arbre seul ; (2) donne l'importance des variables et demande peu de réglages.
**Limites :** (1) moins interprétable qu'un arbre unique ; (2) plus lente et plus lourde en mémoire quand le nombre d'arbres augmente.

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
**Nom :** Machine à vecteurs de support (*Support Vector Machine*, SVM, ici `SVC` avec noyau RBF). **Famille :** classification supervisée.
**Objectif :** trouver la frontière qui sépare les classes avec la **marge** la plus grande possible, c'est-à-dire en restant le plus loin possible des observations les plus proches.
**Intuition :** parmi toutes les frontières possibles, on choisit celle qui laisse le plus d'espace de chaque côté ; seules les observations proches de la frontière (les vecteurs de support) la déterminent. Avec un noyau RBF, la frontière peut être courbe.
**Étapes :** standardisation → recherche des vecteurs de support en maximisant la marge (avec tolérance aux erreurs contrôlée par C) → calcul du score f(x) pour une nouvelle observation → classe selon le signe de f(x).
**Formule :** f(x) = Σᵢ αᵢ yᵢ K(xᵢ, x) + b, avec xᵢ les vecteurs de support, yᵢ ∈ {−1, +1} leur classe, αᵢ leurs poids, b le biais et K(xᵢ, x) = exp(−γ‖xᵢ − x‖²) le noyau RBF, où γ règle la portée du noyau. La marge vaut 2/‖w‖ dans le cas linéaire.
**Paramètres :** les αᵢ, les vecteurs de support et b (appris). **Hyperparamètres :** C (tolérance aux erreurs, `C=1.0`), le noyau (`rbf`) et γ (`gamma`, valeur par défaut).
**Avantages :** (1) efficace en grande dimension et avec peu d'observations ; (2) frontières non linéaires grâce aux noyaux.
**Limites :** (1) sensible à l'échelle des variables et au choix de C et γ ; (2) lent sur de très grands jeux de données et peu lisible, sans probabilités directes.

Les cercles verts entourent les vecteurs de support : ce sont les observations situées sur la marge ou à l'intérieur, et elles seules déterminent la frontière. Les classes se chevauchant, de nombreux points sont des vecteurs de support. *Le schéma n'utilise que 2 des 6 variables pour pouvoir tracer en 2D.*

### Contextes d'utilisation

- **Cas peu approprié :** très grands jeux de données (centaines de milliers d'observations) : l'entraînement devient long ; un modèle linéaire ou un boosting est plus adapté. Il convient aussi mal quand il faut expliquer la décision.
- **Métriques :** F1-score, précision, rappel, ROC-AUC ou PR-AUC ; surveiller le nombre de vecteurs de support.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation
**Nom :** Gradient Boosting pour la classification (*Gradient Boosting Classifier*, ici `HistGradientBoostingClassifier`). **Famille :** classification supervisée, méthode d'ensemble (boosting).
**Objectif :** classer une observation en construisant les arbres l'un après l'autre, chacun corrigeant les erreurs du précédent.
**Intuition :** le modèle part d'une prédiction simple, mesure où il se trompe, puis ajoute un petit arbre spécialisé dans ces erreurs ; en répétant, il améliore progressivement sa frontière de décision.
**Étapes :** score initial (log-cotes des classes) → calcul du gradient de la perte logistique (l'erreur restante) → ajustement d'un arbre sur ce gradient → mise à jour du score → répétition M fois → probabilité par sigmoïde, puis classe.
**Formule :** F_m(x) = F_{m−1}(x) + η · h_m(x) et P(y=1|x) = 1 / (1 + e^{−F_M(x)}), avec F_m le score après m étapes, h_m le m-ième arbre entraîné sur le gradient de la perte, η le taux d'apprentissage et M le nombre d'arbres.
**Paramètres :** les arbres h_m (appris). **Hyperparamètres :** nombre d'itérations (`max_iter=80`), taux d'apprentissage η, profondeur des arbres, graine (`random_state=42`).
**Avantages :** (1) très performant sur des données tabulaires, y compris non linéaires ; (2) gère les interactions entre variables sans standardisation.
**Limites :** (1) surapprentissage possible si trop d'itérations ; (2) plusieurs hyperparamètres à régler et moins interprétable qu'un arbre seul.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
iterations = [1, 5, 10, 20, 40, 80, 150, 300]
f1_ent, f1_test = [], []
for n in iterations:
    r = HistGradientBoostingClassifier(max_iter=n, random_state=GRAINE).fit(Xc_ent, yc_ent)
    f1_ent.append(f1_score(yc_ent, r.predict(Xc_ent)))
    f1_test.append(f1_score(yc_test, r.predict(Xc_test)))
plt.plot(iterations, f1_ent, marker="o", label="F1 entraînement")
plt.plot(iterations, f1_test, marker="o", label="F1 test")
plt.xscale("log"); plt.xlabel("Nombre d'itérations (arbres)"); plt.ylabel("F1")
plt.title("Gradient Boosting : le score progresse à chaque arbre ajouté")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** l'axe horizontal donne le nombre d'arbres (échelle logarithmique), l'axe vertical le F1. Les deux courbes montent quand on ajoute des arbres : chaque arbre corrige une partie des erreurs restantes. Le F1 d'entraînement atteint 1,0 vers 80 arbres, tandis que le F1 de test se stabilise autour de 0,94 à 0,95 : l'écart entre les deux courbes indique que le modèle ajuste très bien les données d'entraînement, mais ne gagne plus rien sur les données nouvelles.

- **Données et prérequis :** données tabulaires numériques ou catégorielles encodées ; pas de standardisation nécessaire ; jeu de validation séparé pour régler les hyperparamètres.
- **Cas peu approprié :** très peu de données ou besoin d'une décision explicable règle par règle : un modèle plus simple (régression logistique, arbre unique) est plus stable et plus lisible.
- **Métriques :** F1-score, précision, rappel, ROC-AUC ou PR-AUC ; courbes d'apprentissage entraînement / validation.

### Présentation
**Nom :** AdaBoost (*Adaptive Boosting*). **Famille :** classification supervisée, méthode d'ensemble (boosting).
**Objectif :** construire un classifieur performant en combinant de nombreux classifieurs très simples (par défaut, des arbres de profondeur 1, appelés « souches »).
**Intuition :** après chaque classifieur, on augmente le poids des observations mal classées pour que le suivant se concentre dessus ; les classifieurs les plus précis ont ensuite plus de poids dans le vote final.
**Étapes :** poids égaux pour toutes les observations → entraînement d'une souche → calcul de son erreur pondérée et de son poids α → augmentation du poids des observations mal classées → répétition M fois → vote pondéré des M souches.
**Formule :** H(x) = signe( Σ_{m=1..M} αₘ · hₘ(x) ), avec hₘ le m-ième classifieur simple (sortie −1 ou +1), αₘ = ½ ln((1 − εₘ)/εₘ) son poids et εₘ son erreur pondérée sur les données d'entraînement.
**Paramètres :** les souches hₘ et leurs poids αₘ (appris). **Hyperparamètres :** nombre de classifieurs (`n_estimators=70`), taux d'apprentissage, profondeur du classifieur de base (1 par défaut), graine (`random_state=42`).
**Avantages :** (1) simple et performant avec des classifieurs très faibles ; (2) peu de réglages et pas de standardisation nécessaire.
**Limites :** (1) sensible aux données bruitées et aux valeurs aberrantes, dont le poids ne cesse d'augmenter ; (2) entraînement séquentiel, donc non parallélisable.

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

### Présentation
**Nom :** K-means (*K-Means*, algorithme des k-moyennes). **Famille :** clustering (apprentissage non supervisé).
**Objectif :** regrouper des observations en k groupes homogènes, sans étiquettes connues à l'avance.
**Intuition :** chaque groupe est représenté par son centre (centroïde) ; une observation appartient au groupe dont le centroïde est le plus proche.
**Étapes :** choix de k → initialisation de k centroïdes → affectation de chaque point au centroïde le plus proche → recalcul de chaque centroïde comme moyenne de ses points → répétition des deux dernières étapes jusqu'à stabilité.
**Critère :** on minimise l'inertie J = Σ_{k=1..K} Σ_{x∈C_k} ‖x − μ_k‖², avec C_k le groupe k, μ_k son centroïde (la moyenne de ses points) et ‖·‖ la distance euclidienne.
**Paramètres :** les centroïdes μ_k et l'affectation des points (appris). **Hyperparamètres :** nombre de groupes (`n_clusters=3`), nombre d'initialisations (`n_init=10`), graine (`random_state=42`).
**Avantages :** (1) simple, rapide et facile à interpréter ; (2) adapté à de grands jeux de données.
**Limites :** (1) le nombre de groupes k doit être choisi à l'avance ; (2) suppose des groupes compacts et de forme sphérique, et est sensible aux valeurs extrêmes et à l'échelle des variables.

```python
fig, axs = plt.subplots(1, 2, figsize=(11, 4))
# Panneau 1 : résultat final avec centroïdes
km = KMeans(n_clusters=3, n_init=10, random_state=GRAINE).fit(X_clu)
axs[0].scatter(X_clu[:, 0], X_clu[:, 1], c=km.labels_, s=12, cmap="viridis")
axs[0].scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], c="red", marker="X", s=200, edgecolor="k", label="Centroïdes")
axs[0].set_title("Groupes trouvés et centroïdes (k = 3)")
axs[0].set_xlabel("Variable 1"); axs[0].set_ylabel("Variable 2"); axs[0].legend()
# Panneau 2 : inertie selon k (méthode du coude)
ks = range(1, 8)
inerties = [KMeans(n_clusters=k, n_init=10, random_state=GRAINE).fit(X_clu).inertia_ for k in ks]
axs[1].plot(list(ks), inerties, marker="o")
axs[1].set_xlabel("Nombre de groupes k"); axs[1].set_ylabel("Inertie")
axs[1].set_title("Méthode du coude : choix de k")
plt.show()
```

### Schéma explicatif
**Comment le lire :** à gauche, chaque point est coloré selon le groupe auquel K-means l'a affecté ; les croix rouges sont les centroïdes, c'est-à-dire les moyennes de chaque groupe. À droite, la courbe montre l'inertie (somme des carrés des distances aux centroïdes) en fonction de k : elle diminue toujours quand k augmente, mais la baisse devient beaucoup plus lente après k = 3. Ce « coude » suggère que 3 groupes décrivent bien ces données.

- **Données et prérequis :** variables numériques, standardisées pour que chacune pèse autant ; k à choisir (coude, silhouette) ; peu de valeurs extrêmes.
- **Cas peu approprié :** des groupes de forme allongée ou imbriquée (par exemple deux anneaux concentriques) ou de tailles très différentes : K-means les découpe mal ; DBSCAN ou un mélange gaussien conviennent mieux.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Présentation
**Nom :** Classification hiérarchique ascendante (*Agglomerative Hierarchical Clustering*, CAH). **Famille :** clustering (apprentissage non supervisé).
**Objectif :** regrouper les observations en construisant une hiérarchie de groupes emboîtés, que l'on peut couper au niveau voulu pour obtenir k groupes.
**Intuition :** au départ, chaque point forme son propre groupe ; on fusionne ensuite à chaque étape les deux groupes les plus proches, jusqu'à n'en avoir plus qu'un seul. L'historique des fusions se représente par un dendrogramme.
**Étapes :** chaque point = un groupe → calcul des distances entre groupes → fusion des deux groupes les plus proches → répétition jusqu'à un seul groupe → coupure de l'arbre pour obtenir k groupes.
**Critère (liaison de Ward, utilisée par défaut dans scikit-learn) :** on fusionne les deux groupes A et B dont la fusion augmente le moins la variance interne, Δ(A, B) = (|A|·|B| / (|A|+|B|)) · ‖μ_A − μ_B‖², avec |A| et |B| les effectifs des groupes et μ_A, μ_B leurs centres.
**Paramètres :** la hiérarchie de fusions (calculée). **Hyperparamètres :** nombre de groupes (`n_clusters=3`), type de liaison (`ward` par défaut), distance utilisée.
**Avantages :** (1) pas besoin de fixer k à l'avance pour construire la hiérarchie, et le dendrogramme aide à choisir ; (2) résultat déterministe, sans initialisation aléatoire.
**Limites :** (1) coût de calcul élevé sur de grands jeux de données ; (2) une fusion faite n'est jamais remise en cause, et le résultat dépend du choix de la liaison.

```python
from scipy.cluster.hierarchy import linkage, dendrogram
Z = linkage(X_clu, method="ward")
plt.figure(figsize=(10, 4))
dendrogram(Z, truncate_mode="lastp", p=20, show_leaf_counts=True)
plt.axhline(y=Z[-3, 2] - 1, color="red", linestyle="--", label="Coupure à 3 groupes")
plt.xlabel("Groupes (ou nombre de points entre parenthèses)")
plt.ylabel("Distance de fusion (Ward)")
plt.title("Dendrogramme : historique des fusions")
plt.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** en bas, les petites branches sont des groupes déjà fusionnés (le nombre entre parenthèses indique combien de points ils contiennent). En montant, des groupes sont fusionnés ; la hauteur d'une barre horizontale est la distance à laquelle la fusion a lieu. Les fusions du bas se font à de petites distances (≈ 8 à 9), alors que les deux dernières fusions se font à des distances beaucoup plus grandes (≈ 90 puis ≈ 152). La ligne rouge coupe l'arbre juste sous ce grand saut : elle croise 3 branches, donc 3 groupes. *La grande distance entre les dernières fusions est le signe de groupes bien séparés.*

### Contextes d'utilisation
- **Cas 1 (marketing) :** construire une typologie de clients et visualiser comment les segments se rapprochent les uns des autres, du plus fin au plus global.
- **Cas 2 (commerce international) :** regrouper des pays selon leurs profils d'échanges ou d'indicateurs économiques, avec un dendrogramme facile à présenter.
- **Données et prérequis :** variables numériques standardisées, jeu de données de taille petite à moyenne (la matrice de distances croît avec le carré du nombre d'observations).
- **Cas peu approprié :** très grands jeux de données (plusieurs dizaines de milliers de points) : le temps de calcul et la mémoire deviennent trop importants ; K-means est alors préférable.
- **Métriques :** coefficient de silhouette, hauteur des fusions dans le dendrogramme, stabilité selon la liaison ; indice de Rand ajusté si des étiquettes de référence existent (évaluation externe uniquement).

### Analyse de l'expérience Python
- **Entrées :** les mêmes 240 points synthétiques en 2 dimensions que pour K-means (`make_blobs`, 3 groupes de 80 points), sans découpage entraînement / test.
- **Hyperparamètres :** `n_clusters=3`, liaison de Ward par défaut.
- **Sortie :** coefficient de silhouette ≈ 0,87, identique à celui de K-means.
- **Ce qu'on peut conclure :** les deux méthodes retrouvent ici le même découpage en trois groupes, ce qui s'explique par des groupes compacts et très bien séparés. La silhouette proche de 1 indique que chaque point est bien plus proche de son groupe que des autres.
- **Ce qu'on ne peut pas conclure :** que les deux méthodes donnent toujours le même résultat : elles peuvent différer sur des groupes allongés, de tailles inégales ou bruités. Les données sont synthétiques et constituent le cas le plus favorable ; la hiérarchie n'a pas été testée avec d'autres types de liaison.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Présentation
**Nom :** DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*, regroupement spatial fondé sur la densité). **Famille :** clustering (apprentissage non supervisé), avec détection de bruit.
**Objectif :** former des groupes à partir des zones denses de points, sans fixer le nombre de groupes, et repérer les points isolés comme bruit.
**Intuition :** un groupe est une région où les points sont serrés les uns contre les autres ; les points situés dans des zones peu denses sont considérés comme du bruit.
**Étapes :** pour chaque point, on compte ses voisins dans un rayon eps → s'il en a au moins `min_samples`, c'est un point central → les points centraux voisins sont reliés en un même groupe → les points proches d'un point central rejoignent son groupe (points de bordure) → les autres sont du bruit (étiquette −1).
**Critère :** un point x est central si |{x' : d(x, x') ≤ ε}| ≥ m, avec d la distance euclidienne, ε le rayon de voisinage et m le nombre minimal de voisins (`min_samples`).
**Paramètres :** aucun paramètre appris ; l'algorithme calcule directement les groupes. **Hyperparamètres :** ε (`eps=0.8`) et `min_samples=5`.
**Avantages :** (1) détecte des groupes de forme quelconque et ne demande pas k ; (2) identifie explicitement le bruit.
**Limites :** (1) très sensible au choix de ε et de `min_samples` ; (2) difficile quand les groupes ont des densités très différentes, et peu efficace en grande dimension.

```python
from sklearn.cluster import DBSCAN
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
for ax, eps in zip(axs, [0.3, 0.8, 2.0]):
    lab = DBSCAN(eps=eps, min_samples=5).fit_predict(X_clu)
    bruit = lab == -1
    ax.scatter(X_clu[~bruit, 0], X_clu[~bruit, 1], c=lab[~bruit], s=12, cmap="viridis")
    ax.scatter(X_clu[bruit, 0], X_clu[bruit, 1], c="red", marker="x", s=30, label="Bruit")
    nb_groupes = len(set(lab)) - (1 if -1 in lab else 0)
    ax.set_title(f"eps = {eps} : {nb_groupes} groupe(s), {bruit.sum()} bruit")
    ax.set_xlabel("Variable 1"); ax.set_ylabel("Variable 2"); ax.legend()
plt.suptitle("DBSCAN : effet du rayon eps (min_samples = 5)")
plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque panneau montre le résultat de DBSCAN pour une valeur différente du rayon eps. Les points colorés appartiennent à un groupe ; les croix rouges sont les points de bruit. Avec eps = 0,3 (trop petit), peu de points ont assez de voisins : on obtient 9 petits groupes et 138 points de bruit. Avec eps = 0,8 (valeur de l'exemple), on retrouve les 3 groupes et seulement 8 points de bruit, situés en périphérie. Avec eps = 2,0 (plus grand), il n'y a plus aucun bruit : tous les points sont absorbés dans les 3 groupes. Ce schéma montre que le résultat dépend fortement de eps.

### Contextes d'utilisation
- **Cas 1 (logistique) :** repérer des zones de forte concentration de livraisons ou de pannes sur une carte, et isoler les points aberrants.
- **Cas 2 (finance) :** détecter des transactions isolées qui ne ressemblent à aucun groupe habituel (le bruit sert de premier signal d'alerte).
- **Données et prérequis :** variables numériques standardisées, densité à peu près homogène entre les groupes, choix de eps et `min_samples` guidé par le graphique des distances aux k plus proches voisins.
- **Cas peu approprié :** des groupes de densités très différentes ou des données en grande dimension : une seule valeur de eps ne convient pas à tous les groupes, et les distances perdent leur sens.
- **Métriques :** nombre de groupes et de points de bruit, coefficient de silhouette calculé hors bruit, stabilité selon eps ; indice de Rand ajusté si des étiquettes de référence existent (évaluation externe uniquement).

### Analyse de l'expérience Python
- **Entrées :** les mêmes 240 points synthétiques en 2 dimensions que pour K-means et la classification hiérarchique, sans découpage entraînement / test.
- **Hyperparamètres :** `eps=0.8`, `min_samples=5`.
- **Sortie :** 3 groupes (étiquettes 0, 1, 2) et 8 points de bruit (étiquette −1). Les groupes contiennent respectivement 76, 79 et 77 points.
- **Ce qu'on peut conclure :** avec ces paramètres, DBSCAN retrouve les 3 groupes de l'exemple et met à part 8 points éloignés des zones denses. Contrairement à K-means, ces points ne sont pas forcés dans un groupe.
- **Ce qu'on ne peut pas conclure :** que ces 8 points sont de vraies anomalies : ils sont simplement en périphérie des groupes, générés comme les autres. Le résultat dépend de eps (voir le schéma), qui n'a pas été optimisé, et les données synthétiques sont le cas le plus favorable. Aucune métrique de qualité n'est calculée dans la cellule fournie.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Présentation
**Nom :** Mélanges gaussiens (*Gaussian Mixture Models*, GMM). **Famille :** clustering probabiliste (apprentissage non supervisé).
**Objectif :** modéliser les données comme un mélange de plusieurs lois normales, et attribuer à chaque point une probabilité d'appartenance à chaque groupe.
**Intuition :** on suppose que chaque groupe suit une loi gaussienne (une « cloche » en forme d'ellipse) ; chaque point provient de l'une d'elles, mais on ne sait pas laquelle. L'algorithme estime les gaussiennes les plus plausibles.
**Étapes (algorithme EM) :** initialisation des gaussiennes → étape E : calcul, pour chaque point, de la probabilité d'appartenance à chaque gaussienne → étape M : mise à jour des poids, moyennes et covariances à partir de ces probabilités → répétition jusqu'à convergence.
**Formule :** p(x) = Σ_{k=1..K} πₖ · N(x | μₖ, Σₖ), avec K le nombre de composantes, πₖ le poids du groupe k (Σπₖ = 1), μₖ sa moyenne, Σₖ sa matrice de covariance et N la densité de la loi normale.
**Paramètres :** πₖ, μₖ et Σₖ (appris). **Hyperparamètres :** nombre de composantes (`n_components=3`), type de covariance (`full` par défaut), graine (`random_state=42`).
**Avantages :** (1) donne des probabilités d'appartenance (affectation « douce ») ; (2) gère des groupes de forme elliptique et de tailles différentes.
**Limites :** (1) suppose des groupes de forme gaussienne et le nombre de composantes doit être choisi ; (2) sensible à l'initialisation et peut converger vers un optimum local.

```python
from sklearn.mixture import GaussianMixture
from matplotlib.patches import Ellipse
gmm = GaussianMixture(n_components=3, random_state=GRAINE).fit(X_clu)
etiq = gmm.predict(X_clu)
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(X_clu[:, 0], X_clu[:, 1], c=etiq, s=12, cmap="viridis")
for mu, cov in zip(gmm.means_, gmm.covariances_):
    val, vec = np.linalg.eigh(cov)
    angle = np.degrees(np.arctan2(vec[1, 0], vec[0, 0]))
    for n in [1, 2]:  # ellipses à 1 et 2 écarts-types
        ax.add_patch(Ellipse(mu, 2*n*np.sqrt(val[0]), 2*n*np.sqrt(val[1]), angle=angle, fill=False, edgecolor="k"))
ax.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c="red", marker="X", s=150, edgecolor="k", label="Moyennes")
ax.set_xlabel("Variable 1"); ax.set_ylabel("Variable 2")
ax.set_title("Mélange gaussien : 3 gaussiennes (ellipses à 1 et 2 écarts-types)")
ax.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque point est coloré selon la gaussienne la plus probable pour lui. Chaque paire d'ellipses noires représente une gaussienne : l'ellipse intérieure couvre environ 1 écart-type autour de la moyenne (croix rouge), l'ellipse extérieure environ 2 écarts-types. Sa forme montre la dispersion du groupe dans chaque direction : ici, les ellipses sont presque des cercles, car les groupes ont été générés ainsi. Un point situé entre deux ellipses aurait des probabilités d'appartenance partagées entre deux groupes ; ici, les groupes sont si éloignés que ce n'est pas le cas.

### Contextes d'utilisation
- **Cas 1 (marketing) :** segmenter la clientèle avec des groupes qui se recouvrent, en obtenant pour chaque client une probabilité d'appartenance à chaque segment.
- **Cas 2 (finance) :** modéliser la distribution des rendements d'un actif comme un mélange de régimes (calme, turbulent) et détecter les observations peu probables.
- **Données et prérequis :** variables numériques continues, groupes approximativement gaussiens, choix du nombre de composantes par un critère comme le BIC.
- **Cas peu approprié :** des groupes de forme non convexe (par exemple en anneau) ou des variables catégorielles : l'hypothèse gaussienne n'est pas respectée ; DBSCAN ou un autre modèle convient mieux.
- **Métriques :** BIC / AIC pour choisir le nombre de composantes, vraisemblance, coefficient de silhouette, stabilité entre initialisations.

### Analyse de l'expérience Python
- **Entrées :** les mêmes 240 points synthétiques en 2 dimensions que pour les algorithmes 17 à 19, sans découpage entraînement / test.
- **Hyperparamètres :** `n_components=3`, `random_state=42`, covariance complète par défaut.
- **Sortie :** poids ≈ 0,333 pour chacune des 3 composantes (soit 80 points sur 240) et silhouette ≈ 0,87.
- **Ce qu'on peut conclure :** le mélange gaussien retrouve les 3 groupes avec des poids égaux, ce qui correspond à la façon dont les données ont été générées. La silhouette est identique à celle de K-means, car les groupes sont compacts et bien séparés. Par ailleurs, le critère BIC, calculé séparément, est le plus bas pour 3 composantes (≈ 1803, contre ≈ 1835 pour 4), ce qui confirme ce choix.
- **Ce qu'on ne peut pas conclure :** que le mélange gaussien est meilleur que K-means : sur ces données, les deux donnent le même résultat. Ses atouts (probabilités d'appartenance, groupes elliptiques) ne se voient pas ici. Les données synthétiques sont le cas le plus favorable : elles ont été générées à partir de gaussiennes.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Présentation
**Nom :** Analyse en composantes principales (*Principal Component Analysis*, ACP / PCA). **Famille :** réduction de dimension (apprentissage non supervisé).
**Objectif :** résumer un grand nombre de variables corrélées par quelques nouvelles variables (les composantes principales), en perdant le moins d'information possible.
**Intuition :** on cherche les directions de l'espace dans lesquelles les données varient le plus ; on projette ensuite les données sur les premières de ces directions.
**Étapes :** standardisation des variables → calcul de la matrice de covariance → recherche de ses vecteurs propres et valeurs propres → classement par valeur propre décroissante → projection des données sur les q premiers vecteurs propres.
**Formule :** z = Wᵀ(x − x̄), avec x l'observation, x̄ la moyenne des observations, W la matrice dont les colonnes sont les q vecteurs propres retenus de la matrice de covariance, et z les coordonnées dans le nouvel espace. La part de variance expliquée par la composante j est λⱼ / Σₖ λₖ, où λⱼ est la j-ième valeur propre.
**Paramètres :** les vecteurs propres (composantes) et les valeurs propres (appris). **Hyperparamètre :** le nombre de composantes conservées (`n_components=2`).
**Avantages :** (1) réduit la dimension et le bruit, et permet de visualiser des données en 2D ; (2) les composantes sont non corrélées et la méthode est rapide.
**Limites :** (1) ne capte que des structures linéaires ; (2) les composantes sont des combinaisons de variables, donc moins faciles à interpréter, et la méthode est sensible à l'échelle des variables.

```python
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
vin = load_wine()
X_vin = StandardScaler().fit_transform(vin.data)
pca_complet = PCA().fit(X_vin)
projection = PCA(n_components=2).fit_transform(X_vin)
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
# Panneau 1 : variance expliquée cumulée
cumul = np.cumsum(pca_complet.explained_variance_ratio_)
axs[0].plot(range(1, len(cumul)+1), cumul, marker="o")
axs[0].axhline(0.9, color="red", linestyle="--", label="90 % de la variance")
axs[0].set_xlabel("Nombre de composantes"); axs[0].set_ylabel("Variance expliquée cumulée")
axs[0].set_title("Combien de composantes garder ?"); axs[0].legend()
# Panneau 2 : projection 2D colorée par cultivar
for classe, nom in enumerate(vin.target_names):
    axs[1].scatter(projection[vin.target == classe, 0], projection[vin.target == classe, 1], s=15, label=nom)
axs[1].set_xlabel("Composante 1"); axs[1].set_ylabel("Composante 2")
axs[1].set_title("Projection sur les 2 premières composantes"); axs[1].legend()
plt.show()
```

### Schéma explicatif
**Comment le lire :** à gauche, la courbe donne la part de variance totale conservée en fonction du nombre de composantes. Les 2 premières composantes en conservent environ 55 %, et il en faut 8 pour dépasser 90 %. À droite, chaque point est un vin projeté sur les 2 premières composantes ; la couleur indique son cultivar, mais l'ACP ne l'a jamais utilisé. On observe que les trois cultivars se regroupent en zones largement séparées : les deux premières composantes captent donc une structure liée aux classes, même si 45 % de la variance est perdue.

### Contextes d'utilisation
- **Cas 1 (finance) :** résumer des dizaines d'indicateurs de marché corrélés en quelques facteurs (par exemple niveau, pente et courbure de la courbe des taux).
- **Cas 2 (marketing) :** synthétiser de nombreuses réponses d'enquête en quelques axes pour visualiser les profils de clients.
- **Données et prérequis :** variables numériques, standardisées avant l'ACP, avec des corrélations entre variables (sinon la réduction ne sert à rien).
- **Cas peu approprié :** des structures non linéaires (par exemple des données en spirale) ou des variables catégorielles : l'ACP les résume mal ; t-SNE ou UMAP conviennent mieux pour visualiser.
- **Métriques :** variance expliquée (par composante et cumulée), erreur de reconstruction ; les corrélations entre variables et composantes aident à les interpréter.

### Analyse de l'expérience Python
- **Entrées :** le jeu `load_wine` de scikit-learn (178 vins, 13 variables chimiques), standardisé par `StandardScaler`. Il n'y a pas de découpage entraînement / test, car la cellule sert à explorer et à visualiser les données.
- **Hyperparamètres :** `n_components=2`.
- **Sortie :** variance expliquée ≈ 0,362 pour la 1re composante et ≈ 0,192 pour la 2e, soit ≈ 0,554 au total.
- **Ce qu'on peut conclure :** les deux premières composantes résument environ 55 % de l'information des 13 variables, ce qui permet de représenter les vins en 2D. Les 45 % restants sont perdus dans cette représentation.
- **Ce qu'on ne peut pas conclure :** que la projection à 2 dimensions suffit à tout expliquer, ni que les composantes ont un sens chimique précis : il faudrait examiner les coefficients de chaque composante. Le nuage de points de la cellule fournie ne colore pas les classes : on ne peut donc pas en tirer de conclusion sur les cultivars à partir de cette seule cellule. Le jeu de données est petit et déjà très étudié, il ne remplace pas une étude sur des données réelles de la partie B.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Présentation
**Nom :** t-SNE (*t-distributed Stochastic Neighbor Embedding*). **Famille :** réduction de dimension non linéaire (apprentissage non supervisé), surtout utilisée pour la visualisation.
**Objectif :** représenter des données de grande dimension en 2D (ou 3D) en gardant proches les points qui étaient voisins dans l'espace d'origine.
**Intuition :** on mesure la proximité entre points dans l'espace d'origine sous forme de probabilités, puis on place les points en 2D de façon que ces probabilités soient les plus semblables possible.
**Étapes :** calcul des similarités entre points dans l'espace d'origine (loi gaussienne, réglée par la perplexité) → initialisation aléatoire des points en 2D → calcul des similarités en 2D (loi de Student) → déplacement des points par descente de gradient pour minimiser la divergence entre les deux jeux de similarités.
**Critère :** on minimise la divergence de Kullback-Leibler KL(P‖Q) = Σᵢⱼ pᵢⱼ log(pᵢⱼ / qᵢⱼ), avec pᵢⱼ les similarités entre les points i et j dans l'espace d'origine et qᵢⱼ celles dans l'espace réduit.
**Paramètres :** les coordonnées 2D de chaque point (apprises). **Hyperparamètres :** perplexité (`perplexity=20`, nombre approximatif de voisins considérés), nombre d'itérations, taux d'apprentissage, graine (`random_state=42`).
**Avantages :** (1) très bon pour révéler des groupes locaux dans un nuage de grande dimension ; (2) capte des structures non linéaires.
**Limites :** (1) les distances entre groupes et la taille des groupes n'ont pas de sens fiable ; (2) résultat dépendant de la perplexité et de la graine, et pas de transformation applicable à de nouvelles données.

```python
from sklearn.manifold import TSNE, trustworthiness
from sklearn.decomposition import PCA
vin = load_wine()
X_vin = StandardScaler().fit_transform(vin.data)
perplexites = [5, 20, 50]
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
for ax, p in zip(axs, perplexites):
    proj = TSNE(n_components=2, perplexity=p, random_state=GRAINE).fit_transform(X_vin)
    for classe, nom in enumerate(vin.target_names):
        ax.scatter(proj[vin.target == classe, 0], proj[vin.target == classe, 1], s=12, label=nom)
    score = trustworthiness(X_vin, proj, n_neighbors=10)
    ax.set_title(f"perplexité = {p}\nconservation des voisinages = {score:.3f}")
    ax.set_xlabel("Dimension 1"); ax.set_ylabel("Dimension 2")
axs[0].legend()
plt.suptitle("t-SNE : effet de la perplexité (couleurs = cultivars, non utilisés par t-SNE)")
plt.show()
score_pca = trustworthiness(X_vin, PCA(n_components=2).fit_transform(X_vin), n_neighbors=10)
print("Conservation des voisinages — ACP (2 composantes) :", round(score_pca, 3))
```

### Schéma explicatif
**Comment le lire :** chaque panneau montre le résultat de t-SNE pour une perplexité différente (5, 20 et 50). Chaque point est un vin ; la couleur indique son cultivar, mais t-SNE ne l'a pas utilisé. Les trois cultivars forment des groupes visibles dans les trois panneaux, avec une forme qui change selon la perplexité. Le score affiché en titre mesure la conservation des voisinages : plus il est proche de 1, mieux les voisins d'origine sont préservés. Il est d'environ 0,94 à 0,96 pour t-SNE, contre environ 0,89 pour l'ACP à 2 composantes. *Les distances entre groupes et la taille des groupes ne doivent pas être interprétées.*

### Contextes d'utilisation
- **Cas 1 (marketing) :** visualiser des profils de clients décrits par des dizaines de variables pour repérer des segments naturels avant de faire un clustering.
- **Cas 2 (finance) :** explorer visuellement des transactions ou des entreprises décrites par de nombreux indicateurs, afin de repérer des familles ou des cas atypiques.
- **Données et prérequis :** variables numériques standardisées, jeu de données de taille petite à moyenne (le calcul est coûteux), perplexité inférieure au nombre d'observations.
- **Cas peu approprié :** réduire la dimension pour alimenter un modèle prédictif ou transformer de nouvelles observations (t-SNE n'a pas de transformation réutilisable) ; ne pas non plus l'utiliser pour interpréter des distances entre groupes.
- **Métriques :** conservation des voisinages (`trustworthiness`), divergence KL finale ; l'inspection visuelle reste importante, mais doit être complétée par ces mesures.

### Analyse de l'expérience Python
- **Entrées :** le jeu `load_wine` de scikit-learn (178 vins, 13 variables chimiques), standardisé par `StandardScaler`. Pas de découpage entraînement / test : la cellule sert à visualiser.
- **Hyperparamètres :** `n_components=2`, `perplexity=20`, `random_state=42`.
- **Sortie :** un nuage de points en 2D. La cellule fournie n'affiche aucune valeur numérique et ne colore pas les classes.
- **Ce qu'on peut conclure :** la projection fait apparaître des zones de points regroupés, ce qui suggère une structure locale dans les données. Grâce au calcul supplémentaire de la cellule de schéma, on observe que t-SNE conserve mieux les voisinages (≈ 0,96 pour une perplexité de 20) que l'ACP à 2 composantes (≈ 0,89).
- **Ce qu'on ne peut pas conclure :** que les groupes visibles sont de « vrais » groupes, ni que les distances entre eux ont un sens : t-SNE déforme les distances globales. Le résultat dépend de la perplexité et de la graine, et la cellule fournie ne contient aucune métrique : seules celles que vous avez ajoutées permettent d'argumenter. Le jeu est petit et déjà très étudié, il ne remplace pas les données réelles de la partie B.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

### Présentation
**Nom :** UMAP (*Uniform Manifold Approximation and Projection*). **Famille :** réduction de dimension non linéaire (apprentissage non supervisé), utilisée pour la visualisation et comme prétraitement.
**Objectif :** représenter des données de grande dimension en 2D (ou plus) en conservant à la fois la structure locale (les voisins proches) et une part de la structure globale.
**Intuition :** on suppose que les données se situent sur une « surface » de plus faible dimension plongée dans l'espace d'origine. UMAP construit un graphe de voisinage pondéré entre les points, puis cherche une disposition en 2D dont le graphe soit le plus semblable possible.
**Étapes :** recherche des plus proches voisins de chaque point → construction d'un graphe pondéré (poids = proximité) → initialisation de la disposition en 2D → optimisation de la disposition pour rapprocher les points liés et écarter les autres.
**Critère :** on minimise une entropie croisée entre le graphe d'origine et le graphe en 2D, C = Σ_{(i,j)} [ μᵢⱼ log(μᵢⱼ / νᵢⱼ) + (1 − μᵢⱼ) log((1 − μᵢⱼ) / (1 − νᵢⱼ)) ], avec μᵢⱼ le poids de l'arête entre i et j dans l'espace d'origine et νᵢⱼ celui dans l'espace réduit.
**Paramètres :** les coordonnées 2D de chaque point (apprises). **Hyperparamètres :** nombre de voisins (`n_neighbors=15`), distance minimale entre points (`min_dist`), nombre de dimensions (`n_components=2`), graine (`random_state=42`).
**Avantages :** (1) plus rapide que t-SNE sur de grands jeux de données ; (2) préserve mieux la structure globale que t-SNE et fournit une transformation applicable à de nouvelles données.
**Limites :** (1) résultat sensible à `n_neighbors` et `min_dist`, et les distances entre groupes restent à interpréter avec prudence ; (2) méthode moins transparente mathématiquement que l'ACP, et fixer `random_state` désactive le calcul parallèle (donc plus lent).

```python
import umap
from sklearn.manifold import trustworthiness
vin = load_wine()
X_vin = StandardScaler().fit_transform(vin.data)
voisins = [5, 15, 50]
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
for ax, k in zip(axs, voisins):
    proj = umap.UMAP(n_components=2, n_neighbors=k, random_state=GRAINE).fit_transform(X_vin)
    for classe, nom in enumerate(vin.target_names):
        ax.scatter(proj[vin.target == classe, 0], proj[vin.target == classe, 1], s=12, label=nom)
    score = trustworthiness(X_vin, proj, n_neighbors=10)
    ax.set_title(f"n_neighbors = {k}\nconservation des voisinages = {score:.3f}")
    ax.set_xlabel("Dimension 1"); ax.set_ylabel("Dimension 2")
axs[0].legend()
plt.suptitle("UMAP : effet de n_neighbors (couleurs = cultivars, non utilisés par UMAP)")
plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque panneau montre le résultat d'UMAP pour une valeur différente de `n_neighbors` (5, 15 et 50). Chaque point est un vin ; la couleur indique son cultivar, mais UMAP ne l'a pas utilisé. Avec peu de voisins (5), UMAP met l'accent sur la structure très locale et tend à fragmenter les groupes ; avec davantage de voisins (50), il tient davantage compte de la structure d'ensemble. Le score affiché en titre mesure la conservation des voisinages : plus il est proche de 1, mieux les voisins d'origine sont préservés. D'après mon graphique, ___ (décrivez ce que vous voyez : les 3 cultivars sont-ils séparés ? les groupes changent-ils de forme ?). *Les distances entre groupes et leur taille ne doivent pas être sur-interprétées.*

- **Cas 2 (finance) :** projeter des milliers de transactions ou d'entreprises en 2D pour explorer des familles de comportements, ou réduire la dimension avant un clustering.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

### Présentation
**Nom :** Forêt d'isolement (*Isolation Forest*). **Famille :** détection d'anomalies (apprentissage non supervisé).
**Objectif :** repérer les observations atypiques, c'est-à-dire celles qui diffèrent fortement de la majorité des données.
**Intuition :** une observation rare et éloignée des autres est facile à « isoler » : il suffit de quelques coupures aléatoires pour la séparer du reste. Une observation typique, au milieu des autres, demande beaucoup plus de coupures.
**Étapes :** construction de nombreux arbres où chaque nœud choisit une variable et un seuil au hasard → pour chaque observation, mesure de la profondeur à laquelle elle est isolée → moyenne sur tous les arbres → conversion en score d'anomalie → étiquette −1 (atypique) si le score dépasse un seuil fixé par le taux de contamination.
**Formule :** s(x, n) = 2^{ −E[h(x)] / c(n) }, avec h(x) la profondeur à laquelle x est isolée dans un arbre, E[h(x)] sa moyenne sur tous les arbres, n le nombre d'observations utilisées pour construire chaque arbre et c(n) la profondeur moyenne attendue d'une recherche dans un arbre de n observations (facteur de normalisation). Un score proche de 1 indique une anomalie, proche de 0,5 ou moins une observation normale. *Dans scikit-learn, la méthode `score_samples` renvoie l'opposé de ce score : plus il est bas (négatif), plus l'observation est atypique.*
**Paramètres :** les coupures aléatoires des arbres (construites). **Hyperparamètres :** nombre d'arbres (`n_estimators`, 100 par défaut), taux de contamination (`contamination=0.02`), nombre d'observations par arbre (`max_samples`), graine (`random_state=42`).
**Avantages :** (1) rapide, même sur de grands jeux de données et en grande dimension ; (2) ne demande ni étiquettes ni hypothèse sur la forme des données.
**Limites :** (1) le seuil dépend du paramètre `contamination` qu'il faut choisir à l'avance ; (2) a du mal avec des anomalies proches de groupes normaux ou avec des anomalies groupées.

```python
from sklearn.ensemble import IsolationForest
iso = IsolationForest(contamination=0.02, random_state=GRAINE).fit(X_anomal)
scores = iso.score_samples(X_anomal)       # plus bas = plus atypique
alertes = iso.predict(X_anomal) == -1
fig, ax = plt.subplots(figsize=(7, 5))
points = ax.scatter(X_anomal[:, 0], X_anomal[:, 1], c=scores, cmap="viridis", s=15)
plt.colorbar(points, label="Score (bas = atypique)")
ax.scatter(X_anomal[alertes, 0], X_anomal[alertes, 1], facecolors="none", edgecolors="red", s=140, linewidths=2, label="Alertes")
ax.scatter(X_anomal[-2:, 0], X_anomal[-2:, 1], marker="*", c="orange", edgecolor="k", s=220, label="2 points ajoutés à la main")
ax.set_xlabel("Variable 1"); ax.set_ylabel("Variable 2")
ax.set_title("Isolation Forest : score d'anomalie par point")
ax.legend(); plt.show()
```

### Schéma explicatif
**Comment le lire :** chaque point est coloré selon son score d'anomalie : les couleurs sombres (scores bas) indiquent les points les plus atypiques, les couleurs claires les points typiques. Les cercles rouges entourent les points signalés par le modèle (étiquette −1). Les étoiles orange sont les deux points ajoutés à la main. Le point [9, 9], isolé loin de tout groupe, a le score le plus bas et est signalé. Le point [-8, -8], lui, se trouve juste à côté d'un des trois groupes : son score est moins extrême et il n'est pas signalé. Les trois groupes montrent ainsi que la distance à la majorité, et non le fait d'être « hors du nuage global », guide le score.

### Contextes d'utilisation
- **Cas 1 (finance) :** repérer des transactions bancaires inhabituelles (montant, heure, lieu) pour déclencher une vérification de fraude.
- **Cas 2 (logistique / gestion) :** détecter des commandes ou des expéditions dont les délais, poids ou coûts sortent nettement de la norme, pour contrôler la qualité des données ou des opérations.
- **Données et prérequis :** variables numériques, anomalies rares par rapport aux observations normales, choix du taux de contamination (ou d'un seuil sur le score).
- **Cas peu approprié :** des anomalies nombreuses ou très proches des observations normales (comme le point [-8, -8] ici), ou des données où les anomalies forment elles-mêmes un groupe dense : le modèle les considère comme normales.
- **Métriques :** si des étiquettes fiables existent, précision et rappel des alertes ou PR-AUC sur le score ; sinon, inspection manuelle des alertes les plus fortes et stabilité du classement selon la graine. Sans étiquette, la précision et le rappel ne peuvent pas être calculés.

### Analyse de l'expérience Python
- **Entrées :** 242 points en 2 dimensions : les 240 points synthétiques de `make_blobs` (3 groupes) et 2 points ajoutés à la main, [9, 9] et [-8, -8]. Pas de découpage entraînement / test : le modèle est appliqué aux données sur lesquelles il est ajusté.
- **Hyperparamètres :** `contamination=0.02`, `random_state=42`, 100 arbres par défaut.
- **Sortie :** 5 observations signalées, d'indices 20, 79, 130, 147 et 240. Cela correspond à 2 % de 242 points, arrondi à 5, puisque `contamination` fixe la proportion d'alertes.
- **Ce qu'on peut conclure :** le point [9, 9] (indice 240) est bien détecté et reçoit le score le plus bas. En revanche, le point [-8, -8] (indice 241) n'est pas signalé : il est proche de l'un des trois groupes, donc facile à confondre avec une observation normale. Les 4 autres alertes (indices 20, 79, 130, 147) sont des points des groupes, situés en périphérie.
- **Ce qu'on ne peut pas conclure :** que les 4 autres alertes sont de « vraies » anomalies : ce sont des points générés comme les autres, simplement plus éloignés du centre de leur groupe. Le nombre d'alertes est imposé par `contamination`, pas découvert par le modèle. Aucune métrique de type précision ou rappel n'est calculée dans la cellule fournie : seules les 2 anomalies ajoutées sont connues, ce qui ne permet pas une évaluation fiable.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Présentation
**Nom :** Bandits manchots (*Multi-Armed Bandits*). **Famille :** apprentissage par renforcement (cas à un seul état).
**Objectif :** choisir, à chaque tour, parmi plusieurs actions (les « bras ») dont les gains sont aléatoires et inconnus, afin de maximiser la récompense cumulée.
**Intuition :** comme face à plusieurs machines à sous, on ne sait pas laquelle est la meilleure ; il faut jouer pour l'apprendre, mais chaque coup joué sur une mauvaise machine coûte. C'est le dilemme **exploration / exploitation**.
**Étapes :** à chaque tour t → choix d'un bras selon une stratégie → observation de la récompense → mise à jour de l'estimation de la valeur de ce bras. *Dans la cellule fournie, la stratégie est simplement d'alterner les bras (t modulo 3), sans exploitation.*
**Formule :** V̂ₐ = (1/Nₐ) Σ rₜ sur les tours où le bras a a été joué, avec Nₐ le nombre de fois où a a été joué et rₜ la récompense obtenue au tour t (0 ou 1 ici). Le regret cumulé est R_T = T·μ* − Σₜ rₜ, avec μ* la probabilité de gain du meilleur bras et T le nombre de tours.
**Paramètres :** les estimations V̂ₐ (apprises). **Hyperparamètres :** la stratégie de choix (ici alternance), le nombre de tours (500) et la graine (`GRAINE = 42`).
**Avantages :** (1) modèle simple pour formaliser le compromis exploration / exploitation ; (2) applicable à la publicité en ligne, aux tests A/B adaptatifs ou aux recommandations.
**Limites :** (1) ne tient pas compte du contexte (un seul état) ; (2) suppose des gains stables dans le temps et indépendants d'un tour à l'autre.

```python
# Reprend la simulation de la cellule 25, en gardant l'historique des estimations
generateur = np.random.default_rng(GRAINE)
vraies_probas = np.array([.2, .5, .7]); somme = np.zeros(3); comptes = np.zeros(3)
historique = []
for t in range(500):
    bras = t % 3
    gain = int(generateur.random() < vraies_probas[bras])
    somme[bras] += gain; comptes[bras] += 1
    historique.append(somme / np.maximum(comptes, 1))
historique = np.array(historique)
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
pos = np.arange(3); largeur = 0.35
axs[0].bar(pos - largeur/2, vraies_probas, width=largeur, label="Vraie probabilité")
axs[0].bar(pos + largeur/2, somme / comptes, width=largeur, label="Estimation après 500 tours")
axs[0].set_xticks(pos, ["Bras 0", "Bras 1", "Bras 2"]); axs[0].set_ylabel("Probabilité de gain")
axs[0].set_title("Vraies valeurs et valeurs estimées"); axs[0].legend()
for b in range(3):
    axs[1].plot(historique[:, b], label=f"Bras {b}")
    axs[1].axhline(vraies_probas[b], linestyle="--", color=f"C{b}", alpha=0.5)
axs[1].set_xlabel("Nombre de tours"); axs[1].set_ylabel("Estimation du taux de gain")
axs[1].set_title("Convergence des estimations (pointillés = vraies valeurs)"); axs[1].legend()
plt.show()
```

### Schéma explicatif
**Comment le lire :** à gauche, pour chaque bras, la première barre est la vraie probabilité de gain (fixée dans le code) et la seconde l'estimation obtenue après 500 tours. À droite, chaque courbe montre comment l'estimation d'un bras évolue au fil des tours ; les traits pointillés sont les vraies valeurs. Au début, les estimations sont très instables (elles reposent sur quelques essais seulement) ; elles se rapprochent progressivement des vraies valeurs quand le nombre d'essais augmente. Les estimations finales (≈ 0,18, 0,54 et 0,72) restent proches des vraies valeurs (0,2, 0,5 et 0,7), mais pas identiques : c'est l'erreur d'échantillonnage.

### Contextes d'utilisation
- **Cas 1 (marketing) :** choisir parmi plusieurs versions d'une publicité ou d'un titre d'e-mail celle qui obtient le plus de clics, en réaffectant l'audience au fur et à mesure.
- **Cas 2 (commerce / gestion) :** tester plusieurs prix ou plusieurs promotions en ligne en limitant la perte due aux mauvaises options.
- **Données et prérequis :** une récompense mesurable à chaque essai (clic, achat, gain), des actions bien définies en nombre limité, des gains stables dans le temps.
- **Cas peu approprié :** quand le contexte compte (profil du client, moment de la journée) ou quand une action change l'état futur : il faut alors des bandits contextuels ou un apprentissage par renforcement complet (Q-learning, par exemple).
- **Métriques :** récompense cumulée, regret cumulé, part de tours joués sur le meilleur bras, variabilité selon la graine.

### Analyse de l'expérience Python
- **Entrées :** une simulation de 3 bras de probabilités de gain 0,2, 0,5 et 0,7, sur 500 tours, tirés avec un générateur initialisé à 42. Il n'y a pas de jeu de données ni de découpage entraînement / test.
- **Hyperparamètres :** nombre de tours = 500, stratégie = alternance des bras (t modulo 3), graine = 42.
- **Sortie :** valeurs estimées ≈ 0,180, 0,539 et 0,723, calculées à partir de 167, 167 et 166 essais par bras.
- **Ce qu'on peut conclure :** avec suffisamment d'essais par bras, les estimations se rapprochent des vraies probabilités et permettent de classer correctement les bras (bras 2 > bras 1 > bras 0).
- **Ce qu'on ne peut pas conclure :** que cette stratégie est efficace pour maximiser les gains : en alternant les bras, un tiers des tours est joué sur le meilleur bras, un autre tiers sur le plus mauvais. Le gain total simulé est de 240 environ, loin des environ 350 attendus si l'on avait toujours joué le meilleur bras. On ne peut pas non plus généraliser d'une seule simulation : d'autres graines donneraient d'autres estimations, et ici les vraies valeurs sont connues, ce qui n'est pas le cas en pratique.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Présentation
**Nom :** Epsilon-greedy (*ε-greedy*, stratégie epsilon-glouton). **Famille :** apprentissage par renforcement (bandits manchots).
**Objectif :** choisir à chaque tour un bras en équilibrant exploration (essayer des bras peu connus) et exploitation (jouer le bras qui semble le meilleur), pour maximiser la récompense cumulée.
**Intuition :** la plupart du temps, on joue le bras qui paraît le meilleur ; de temps en temps, avec une petite probabilité ε, on joue un bras au hasard pour vérifier qu'on ne se trompe pas.
**Étapes :** à chaque tour t → avec probabilité ε, choix d'un bras au hasard ; sinon choix du bras de plus grande valeur estimée → observation de la récompense → mise à jour de l'estimation de ce bras.
**Formule :** aₜ = bras tiré au hasard avec probabilité ε, et aₜ = argmaxₐ V̂ₐ avec probabilité 1 − ε, où V̂ₐ est la valeur estimée du bras a. Dans le code, V̂ₐ = (Sₐ + 1) / (Nₐ + 1), avec Sₐ la somme des gains du bras a et Nₐ le nombre de fois où il a été joué ; ce « +1 » donne aux bras peu essayés une estimation initiale optimiste, qui incite à les tester.
**Paramètres :** les estimations V̂ₐ (apprises). **Hyperparamètres :** ε (`epsilon=0.1`), nombre de tours (1 000) et graine (`GRAINE = 42`).
**Avantages :** (1) très simple à programmer et à expliquer ; (2) garantit que tous les bras continuent d'être essayés.
**Limites :** (1) l'exploration reste aveugle : même quand un bras est clairement mauvais, on continue de le jouer avec probabilité ε/3 ; (2) le choix de ε est délicat et la stratégie ne tient pas compte de l'incertitude sur chaque bras.

```python
vraies_probas = np.array([.2, .5, .7])
def simuler(epsilon, graine, T=1000):
    g = np.random.default_rng(graine); s = np.zeros(3); c = np.zeros(3); cumul = []
    for t in range(T):
        b = int(g.integers(3)) if g.random() < epsilon else int(np.argmax((s + 1) / (c + 1)))
        s[b] += int(g.random() < vraies_probas[b]); c[b] += 1; cumul.append(s.sum())
    return c, np.array(cumul)
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
choix, _ = simuler(0.1, GRAINE)
axs[0].bar(["Bras 0 (p=0,2)", "Bras 1 (p=0,5)", "Bras 2 (p=0,7)"], choix, color=["tab:gray", "tab:gray", "tab:green"])
axs[0].set_ylabel("Nombre de fois choisi"); axs[0].set_title("Epsilon-greedy (ε = 0,1) : choix des bras")
for eps in [0, 0.1, 1.0]:
    c, cumul = simuler(eps, GRAINE)
    axs[1].plot(cumul, label=f"ε = {eps}")
axs[1].set_xlabel("Nombre de tours"); axs[1].set_ylabel("Récompense cumulée")
axs[1].set_title("Récompense cumulée selon ε (graine 42)"); axs[1].legend()
plt.show()
# Variation entre graines (20 graines, 1000 tours)
for eps in [0, 0.1, 1.0]:
    gains = [simuler(eps, graine)[1][-1] for graine in range(20)]
    print(f"ε = {eps} : récompense totale moyenne = {np.mean(gains):.1f}, écart-type = {np.std(gains):.1f}")
```

### Schéma explicatif
**Comment le lire :** à gauche, le nombre de fois où chaque bras a été choisi avec ε = 0,1 : le meilleur bras (vert) concentre la grande majorité des choix, les deux autres ne sont joués que par exploration. À droite, la récompense cumulée au cours du temps pour trois valeurs de ε (graine 42). Avec ε = 1, le modèle choisit toujours au hasard (aucune exploitation) : la courbe est la plus basse. Avec ε = 0 (jamais d'exploration) et ε = 0,1, les courbes sont proches pour cette graine ; la comparaison sur 20 graines, affichée sous la figure, indique la variabilité de ces résultats.

### Contextes d'utilisation
- **Cas 1 (marketing) :** choisir parmi plusieurs bannières publicitaires, en montrant surtout celle qui obtient le plus de clics tout en continuant d'en tester d'autres.
- **Cas 2 (commerce / gestion) :** tester plusieurs prix ou offres promotionnelles en ligne, en limitant la perte due aux options moins bonnes.
- **Données et prérequis :** une récompense mesurable à chaque essai, un petit nombre d'actions, des gains à peu près stables dans le temps.
- **Cas peu approprié :** un environnement où les gains évoluent fortement (tendances, saisonnalité) avec un ε fixe, ou quand une mauvaise action est très coûteuse (santé, sécurité) : l'exploration aveugle y est risquée.
- **Métriques :** récompense cumulée, regret cumulé, part de tours joués sur le meilleur bras, variabilité entre graines.

### Analyse de l'expérience Python
- **Entrées :** une simulation de 3 bras de probabilités de gain 0,2, 0,5 et 0,7, sur 1 000 tours, avec graine 42. Pas de jeu de données ni de découpage entraînement / test.
- **Hyperparamètres :** `epsilon=0.1`, 1 000 tours, graine 42, estimation initiale optimiste (Sₐ + 1) / (Nₐ + 1).
- **Sortie :** 25, 49 et 926 choix pour les bras 0, 1 et 2 ; valeurs estimées ≈ 0,20, 0,67 et 0,69.
- **Ce qu'on peut conclure :** le modèle identifie le meilleur bras et le joue dans environ 93 % des tours, alors que l'alternance de l'algorithme 25 n'en jouait qu'un tiers. L'estimation du bras 1 (≈ 0,67 contre 0,5 en réalité) est trop élevée parce qu'elle repose sur 49 essais seulement : peu d'essais donnent des estimations bruitées.
- **Ce qu'on ne peut pas conclure :** qu'ε = 0,1 est le meilleur réglage, ni qu'un ε plus grand ou plus petit serait moins bon, à partir d'une seule graine. ## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### 1. Présentation

Le **Q-learning (apprentissage par fonction de valeur d'action)** est un algorithme d'apprentissage par renforcement (*Reinforcement Learning*). Son objectif est d'apprendre à un agent à prendre les meilleures décisions dans un environnement afin de maximiser les récompenses cumulées.

Son principe repose sur l'interaction entre un agent et un environnement. À chaque étape, l'agent observe un état, choisit une action, reçoit une récompense et découvre un nouvel état.

L'algorithme utilise une table Q qui associe à chaque couple état-action une valeur représentant la récompense future espérée. Cette table est progressivement mise à jour grâce aux expériences de l'agent.

La formule fondamentale est :

\[
Q(s,a)\leftarrow Q(s,a)+\alpha[r+\gamma\max_{a'}Q(s',a')-Q(s,a)]
\]

Avec :

- \(Q(s,a)\) : valeur actuelle de l'action \(a\) dans l'état \(s\).
- \(\alpha\) : taux d'apprentissage (*learning rate*).
- \(r\) : récompense reçue.
- \(\gamma\) : facteur d'actualisation (*discount factor*).
- \(s'\) : nouvel état.
- \(a'\) : action possible dans le nouvel état.
- \(\max Q(s',a')\) : meilleure valeur future estimée.

Les paramètres appris sont les valeurs de la table Q. Les principaux hyperparamètres sont le taux d'apprentissage, le facteur d'actualisation et la stratégie d'exploration.

**Avantages :**
- Il apprend une politique de décision sans nécessiter un modèle préalable de l'environnement.
- Il est relativement simple à comprendre et à implémenter dans les environnements à états discrets.

**Limites :**
- La table Q devient très volumineuse lorsque le nombre d'états et d'actions augmente.
- L'apprentissage peut être lent et dépend fortement du choix des hyperparamètres et de l'exploration.

Le Q-learning fonctionne selon une boucle d'apprentissage entre l'agent et son environnement :

2. Il choisit une action.
3. L'environnement exécute l'action et renvoie une récompense ainsi qu'un nouvel état.
4. L'agent actualise la valeur Q à l'aide de la formule de Bellman.
5. Le processus se répète jusqu'à la fin de l'épisode.

Le point essentiel du Q-learning est qu'il utilise la meilleure action future possible pour mettre à jour la table Q, même si cette action n'est pas nécessairement celle qui sera réellement choisie ensuite.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### 1. Présentation

Le **SARSA (State-Action-Reward-State-Action)** est un algorithme d'apprentissage par renforcement (*Reinforcement Learning*). Son objectif est d'apprendre à un agent quelle action choisir dans chaque état afin de maximiser les récompenses cumulées.

Son fonctionnement repose sur l'interaction entre l'agent et son environnement. À chaque étape, l'agent observe un état, choisit une action, reçoit une récompense, puis observe le nouvel état et choisit une nouvelle action.

Contrairement au Q-learning, SARSA utilise réellement l'action suivante choisie par la politique de l'agent pour mettre à jour la table Q. Il apprend donc en tenant compte de sa propre stratégie d'exploration.

Sa formule de mise à jour est :

\[
Q(s,a)\leftarrow Q(s,a)+\alpha[r+\gamma Q(s',a')-Q(s,a)]
\]

Avec :

- \(Q(s,a)\) : valeur actuelle du couple état-action.
- \(\alpha\) : taux d'apprentissage.
- \(r\) : récompense reçue.
- \(\gamma\) : facteur d'actualisation.
- \(s'\) : nouvel état.
- \(a'\) : action effectivement choisie dans le nouvel état.

Les valeurs de la table Q constituent les paramètres appris. Les hyperparamètres principaux sont le taux d'apprentissage, le facteur d'actualisation et le taux d'exploration.

**Avantages :**
- Il prend en compte les actions réellement choisies pendant l'exploration.
- Il peut apprendre une politique plus prudente lorsque l'exploration présente des risques.

**Limites :**
- Il peut nécessiter davantage d'expériences pour apprendre une bonne politique.
- Ses performances dépendent de la stratégie d'exploration et du choix des hyperparamètres.

La différence essentielle avec Q-learning réside dans le calcul de la cible :

C'est pourquoi SARSA est considéré comme un algorithme *On-policy* : il apprend la valeur des actions en suivant la politique qu'il utilise réellement.

```python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
```

### 1. Présentation

Le **DQN (Deep Q-Network)** est un algorithme d'apprentissage par renforcement profond (*Deep Reinforcement Learning*). Il combine le Q-learning avec les réseaux de neurones artificiels afin de résoudre des problèmes possédant un grand nombre d'états.

Contrairement au Q-learning classique qui utilise une table Q, DQN utilise un réseau de neurones pour approximer les valeurs des actions. L'agent observe un état, le transmet au réseau neuronal et obtient une estimation des valeurs Q pour les différentes actions possibles.

L'agent choisit ensuite une action, reçoit une récompense et observe un nouvel état. Le réseau est entraîné progressivement pour améliorer ses estimations.

La cible utilisée dans DQN est :

\[
y=r+\gamma(1-d)\max_{a'}Q_{\text{cible}}(s',a')
\]

Avec :

- \(y\) : valeur cible utilisée pour l'apprentissage.
- \(r\) : récompense reçue.
- \(\gamma\) : facteur d'actualisation.
- \(d\) : indicateur de fin d'épisode (1 si terminé, 0 sinon).
- \(s'\) : nouvel état.
- \(a'\) : action possible dans le nouvel état.
- \(Q_{\text{cible}}\) : réseau neuronal cible qui estime les valeurs futures.

Les paramètres appris sont les poids et les biais du réseau neuronal. Les hyperparamètres comprennent le taux d'apprentissage, le facteur d'actualisation, la taille de la mémoire de rejeu et la fréquence de mise à jour du réseau cible.

DQN utilise généralement trois mécanismes importants :
- Un réseau neuronal principal (*Online Network*).
- Un réseau cible (*Target Network*).
- Une mémoire de rejeu (*Experience Replay*).

**Avantages :**
- Il peut traiter des espaces d'états complexes et de grande dimension.
- Il permet d'apprendre des stratégies de décision sans construire une table Q complète.

**Limites :**
- Il nécessite davantage de ressources informatiques et de temps d'apprentissage.
- Son entraînement peut être instable et sensible au choix des hyperparamètres.

Les étapes principales sont :

2. Le réseau neuronal principal estime les valeurs Q des actions possibles.
3. L'agent sélectionne une action.
4. L'environnement retourne une récompense et un nouvel état.
5. L'expérience est stockée dans une mémoire de rejeu.
6. Un mini-lot d'expériences est utilisé pour entraîner le réseau principal.
7. Le réseau cible est périodiquement synchronisé avec le réseau principal.

La différence fondamentale avec le Q-learning classique est que DQN remplace la table Q par un réseau neuronal capable d'approximer les valeurs des actions.

DQN peut être utilisé pour apprendre des stratégies d'allocation d'actifs en fonction de l'évolution des marchés financiers. Le réseau neuronal reçoit des informations comme les prix, les rendements et les indicateurs de risque, puis estime les valeurs des différentes décisions possibles.

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
